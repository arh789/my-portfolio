## 4.1 Local provenance and checkpoint utilities

All generated evidence is append-only within a timestamped private snapshot. This stage preserves a source-only hash of the notebook, package versions, acquisition settings, and stage checkpoints without storing credentials.


````python
TRACKING_PARAMETERS = {
    "utm_source", "utm_medium", "utm_campaign", "utm_term",
    "utm_content", "gclid", "fbclid", "msclkid", "srsltid",
}

class UnsafeTargetError(ValueError):
    pass

class ResponseTooLargeError(ValueError):
    pass

def assert_private_output_path(path):
    root = LOCAL_DATA_ROOT.resolve()
    target = Path(path).resolve()
    if not target.is_relative_to(root):
        raise ValueError(f"Refusing to write outside private output root: {target}")
    return target

def load_secret(environment_name):
    value = os.getenv(environment_name, "").strip()
    if not value:
        raise RuntimeError(f"Missing required environment variable: {environment_name}")
    return value

def stable_id(prefix, value, length=16):
    digest = hashlib.sha256(value.encode("utf-8")).hexdigest()[:length]
    return f"{prefix}_{digest}"

def sha256_bytes(value):
    return hashlib.sha256(value).hexdigest()

def sha256_text(value):
    return sha256_bytes(value.encode("utf-8"))

def normalise_url(url):
    parts = urlsplit((url or "").strip())
    if parts.scheme.lower() not in {"http", "https"} or not parts.hostname:
        return ""
    scheme = parts.scheme.lower()
    hostname = parts.hostname.lower()
    netloc = hostname
    if parts.port and not ((scheme == "http" and parts.port == 80) or (scheme == "https" and parts.port == 443)):
        netloc = f"{hostname}:{parts.port}"
    query = [
        (key, value)
        for key, value in parse_qsl(parts.query, keep_blank_values=True)
        if key.lower() not in TRACKING_PARAMETERS
    ]
    path = re.sub(r"/{2,}", "/", parts.path or "/")
    if path != "/":
        path = path.rstrip("/")
    return urlunsplit((scheme, netloc, path, urlencode(query, doseq=True), ""))

def registered_domain(url):
    result = TLD_EXTRACTOR(url)
    return result.top_domain_under_public_suffix or result.fqdn

def notebook_source_hash(path=NOTEBOOK_PATH):
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"Cannot fingerprint notebook: {path}")
    notebook = json.loads(path.read_text(encoding="utf-8"))
    source_only = [
        {"cell_type": cell.get("cell_type"), "source": cell.get("source", [])}
        for cell in notebook.get("cells", [])
    ]
    return sha256_text(json.dumps(source_only, ensure_ascii=False, sort_keys=True))

def package_versions():
    names = [
        "beautifulsoup4", "lxml", "pandas", "requests", "tldextract", "tqdm",
        "spacy", "scikit-learn", "networkx", "python-louvain", "trafilatura",
        "sentence-transformers", "torch", "transformers", "nbformat", "nbconvert", "matplotlib",
    ]
    versions = {}
    for name in names:
        try:
            versions[name] = importlib_metadata.version(name)
        except importlib_metadata.PackageNotFoundError:
            versions[name] = None
    return versions

def write_json(path, value):
    path = assert_private_output_path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(value, ensure_ascii=False, indent=2), encoding="utf-8")

def write_jsonl(path, records):
    path = assert_private_output_path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8") as handle:
        for record in records:
            handle.write(json.dumps(record, ensure_ascii=False) + "\n")

def append_jsonl(path, record):
    path = assert_private_output_path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("a", encoding="utf-8") as handle:
        handle.write(json.dumps(record, ensure_ascii=False) + "\n")

def read_json(path, default=None):
    path = Path(path)
    return json.loads(path.read_text(encoding="utf-8")) if path.exists() else default

def read_jsonl(path):
    path = Path(path)
    if not path.exists():
        return []
    with path.open(encoding="utf-8") as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_checkpoint(stage, status, **details):
    write_json(CHECKPOINTS_DIR / f"{stage}.json", {
        "stage": stage,
        "status": status,
        "snapshot_id": SNAPSHOT_ID,
        "updated_at": datetime.now(timezone.utc).isoformat(),
        **details,
    })
````

## 4.2 Requests session, target safety, and robots policy

Requests is the explicit transport layer. It retries temporary failures with backoff, identifies itself, rejects non-public network targets, rate-limits every origin request, checks robots policies conservatively, and never follows a redirect without inspecting its target. A missing robots file permits retrieval; an unavailable robots policy defaults to skipping rather than assuming permission.


````python
ROBOTS_CACHE = {}
LAST_REQUEST_BY_DOMAIN = defaultdict(float)

def make_session():
    retry = Retry(
        total=3,
        backoff_factor=0.75,
        status_forcelist=(429, 500, 502, 503, 504),
        allowed_methods=frozenset({"GET", "POST"}),
        respect_retry_after_header=True,
        raise_on_status=False,
    )
    session = requests.Session()
    session.headers.update({"User-Agent": USER_AGENT})
    adapter = HTTPAdapter(max_retries=retry)
    session.mount("https://", adapter)
    session.mount("http://", adapter)
    return session

def assert_public_http_url(url):
    parts = urlsplit(url)
    if parts.scheme.lower() not in {"http", "https"} or not parts.hostname:
        raise UnsafeTargetError(f"Unsupported target URL: {url}")
    port = parts.port or (443 if parts.scheme.lower() == "https" else 80)
    try:
        addresses = {
            item[4][0]
            for item in socket.getaddrinfo(parts.hostname, port, type=socket.SOCK_STREAM)
        }
    except socket.gaierror as exc:
        raise UnsafeTargetError(f"Could not resolve target host: {parts.hostname}") from exc
    if not addresses:
        raise UnsafeTargetError(f"Target host resolved to no addresses: {parts.hostname}")
    unsafe = [address for address in addresses if not ipaddress.ip_address(address).is_global]
    if unsafe:
        raise UnsafeTargetError(f"Refusing non-public target addresses for {parts.hostname}: {unsafe}")
    return url

def redirect_request_url(current_url, location):
    absolute = urljoin(current_url, location)
    parts = urlsplit(absolute)
    if parts.scheme.lower() not in {"http", "https"} or not parts.hostname:
        raise requests.exceptions.InvalidURL(
            f"Invalid redirect target from {current_url}: {location}"
        )
    query = [
        (key, value) for key, value in parse_qsl(parts.query, keep_blank_values=True)
        if key.lower() not in TRACKING_PARAMETERS
    ]
    # Preserve the server-supplied trailing slash while following redirects.
    return urlunsplit((parts.scheme.lower(), parts.netloc.lower(), parts.path or "/", urlencode(query, doseq=True), ""))

def wait_for_domain(url, delay_seconds):
    domain = registered_domain(url) or urlsplit(url).hostname
    elapsed = time.monotonic() - LAST_REQUEST_BY_DOMAIN[domain]
    if elapsed < delay_seconds:
        time.sleep(delay_seconds - elapsed)

def polite_get(session, url, delay_seconds, **kwargs):
    assert_public_http_url(url)
    domain = registered_domain(url) or urlsplit(url).hostname
    wait_for_domain(url, delay_seconds)
    try:
        return session.get(url, timeout=REQUEST_TIMEOUT_SECONDS, **kwargs)
    finally:
        LAST_REQUEST_BY_DOMAIN[domain] = time.monotonic()

def read_limited_response_bytes(response, maximum_bytes):
    declared = response.headers.get("content-length", "").strip()
    if declared.isdigit() and int(declared) > maximum_bytes:
        raise ResponseTooLargeError(
            f"Declared response size {declared} exceeds {maximum_bytes} bytes"
        )
    chunks = []
    total = 0
    for chunk in response.iter_content(chunk_size=65_536):
        if not chunk:
            continue
        total += len(chunk)
        if total > maximum_bytes:
            raise ResponseTooLargeError(
                f"Streamed response exceeded {maximum_bytes} bytes"
            )
        chunks.append(chunk)
    return b"".join(chunks)

def robots_policy(url, session):
    parts = urlsplit(url)
    origin = f"{parts.scheme}://{parts.netloc}"
    if origin in ROBOTS_CACHE:
        return ROBOTS_CACHE[origin]

    robots_url = urljoin(origin, "/robots.txt")
    current_url = robots_url
    redirect_chain = []
    response = None
    try:
        for _ in range(MAX_REDIRECTS + 1):
            response = polite_get(
                session, current_url, REQUEST_DELAY_SECONDS,
                stream=True, allow_redirects=False,
            )
            location = response.headers.get("location")
            if response.status_code in {301, 302, 303, 307, 308} and location:
                next_url = redirect_request_url(current_url, location)
                redirect_chain.append({
                    "from": current_url,
                    "status": response.status_code,
                    "to": next_url,
                })
                response.close()
                response = None
                current_url = next_url
                continue
            break
        else:
            raise requests.TooManyRedirects(f"Too many robots redirects: {robots_url}")

        if response.status_code == 404:
            policy = {
                "parser": None,
                "default_allowed": True,
                "policy_status": "robots_not_found_assumed_allowed",
                "crawl_delay": None,
            }
        else:
            response.raise_for_status()
            raw = read_limited_response_bytes(response, MAX_ROBOTS_BYTES)
            encoding = response.encoding or "utf-8"
            parser = RobotFileParser()
            parser.set_url(current_url)
            parser.parse(raw.decode(encoding, errors="replace").splitlines())
            crawl_delay = parser.crawl_delay(ROBOTS_USER_AGENT)
            if crawl_delay is None:
                crawl_delay = parser.crawl_delay("*")
            policy = {
                "parser": parser,
                "default_allowed": None,
                "policy_status": "robots_policy_available",
                "crawl_delay": crawl_delay,
            }
    except Exception as exc:
        policy = {
            "parser": None,
            "default_allowed": ROBOTS_UNAVAILABLE_POLICY == "allow",
            "policy_status": "robots_unavailable",
            "crawl_delay": None,
            "error": repr(exc),
        }
    finally:
        if response is not None:
            response.close()

    policy.update({
        "robots_url": robots_url,
        "resolved_robots_url": current_url,
        "redirect_chain": redirect_chain,
    })
    ROBOTS_CACHE[origin] = policy
    return policy

def robots_allowed(url, session):
    policy = robots_policy(url, session)
    parser = policy.get("parser")
    allowed = (
        bool(policy.get("default_allowed"))
        if parser is None
        else parser.can_fetch(ROBOTS_USER_AGENT, url)
    )
    reason = (
        "robots_allowed"
        if allowed and parser is not None
        else policy["policy_status"]
        if allowed
        else "explicitly_disallowed"
        if parser is not None
        else "robots_unavailable_skipped"
    )
    return {
        "allowed": allowed,
        "reason": reason,
        "robots_url": policy["robots_url"],
        "crawl_delay": policy.get("crawl_delay"),
        "error": policy.get("error"),
    }
````

## 4.3 Search-result discovery with Requests

The discovery stage preserves the query verbatim, locale, SERP page, visible position, calculated rank, title, snippet, retrieval time, original URL, and normalised candidate URL. It checkpoints after every result page and stores no raw API response or authorization header.


````python
QUERY_HIT_COLUMNS = [
    "snapshot_id", "query_id", "query", "serp_page",
    "page_position", "rank", "serp_url", "candidate_url",
    "serp_title", "serp_snippet", "retrieved_at",
]

def query_id_for(query):
    identity = f"{RUN_NAME}|{GOOGLE_COUNTRY}|{GOOGLE_LANGUAGE}|{query}"
    return stable_id("query", identity)

def save_query_hits(rows):
    frame = pd.DataFrame(rows, columns=QUERY_HIT_COLUMNS)
    if not frame.empty:
        frame = (
            frame.sort_values(["query_id", "candidate_url", "rank"])
            .drop_duplicates(["query_id", "candidate_url"], keep="first")
            .reset_index(drop=True)
        )
    output = assert_private_output_path(QUERY_HITS_FILE)
    output.parent.mkdir(parents=True, exist_ok=True)
    frame.to_csv(output, index=False)
    return frame

def collect_serp_results():
    if ACQUISITION_ACTION not in {"acquire", "resume"}:
        raise RuntimeError("SERP collection is available only in acquire or resume mode.")
    if not NETWORK_APPROVED:
        raise PermissionError("Human approval is required before network acquisition.")
    if not QUERIES or any(not isinstance(query, str) or not query.strip() for query in QUERIES):
        raise ValueError("QUERIES must contain at least one non-empty string.")
    if RESULTS_PER_QUERY < 1:
        raise ValueError("RESULTS_PER_QUERY must be positive.")

    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    CHECKPOINTS_DIR.mkdir(parents=True, exist_ok=True)
    existing = (
        pd.read_csv(QUERY_HITS_FILE).to_dict(orient="records")
        if QUERY_HITS_FILE.exists()
        else []
    )
    rows_by_key = {
        (row["query_id"], row["candidate_url"]): row
        for row in existing
    }
    # Domain dominance is evidence about the SERP, not a collection-time exclusion rule.
    progress = read_json(SERP_PROGRESS_FILE, {"completed_query_ids": []})
    completed_query_ids = set(progress.get("completed_query_ids", []))
    api_key = load_secret("SERPER_API_KEY")
    session = make_session()

    try:
        for query in tqdm(QUERIES, desc=f"Queries: {RUN_NAME}/{SNAPSHOT_ID}"):
            query_id = query_id_for(query)
            if query_id in completed_query_ids:
                continue
            query_rows = [
                row for (row_query_id, _), row in rows_by_key.items()
                if row_query_id == query_id
            ]
            seen_urls = {row["candidate_url"] for row in query_rows}
            page = max((int(row["serp_page"]) for row in query_rows), default=0) + 1
            query_failed = False

            while len(query_rows) < RESULTS_PER_QUERY:
                response = None
                try:
                    response = session.post(
                        "https://google.serper.dev/search",
                        headers={"X-API-KEY": api_key},
                        json={
                            "q": query,
                            "page": page,
                            "gl": GOOGLE_COUNTRY,
                            "hl": GOOGLE_LANGUAGE,
                        },
                        timeout=REQUEST_TIMEOUT_SECONDS,
                    )
                    response.raise_for_status()
                    organic = response.json().get("organic", [])
                except Exception as exc:
                    append_jsonl(SERP_ERRORS_FILE, {
                        "record_type": "serp_error",
                        "status": "api_error",
                        "snapshot_id": SNAPSHOT_ID,
                        "query_id": query_id,
                        "query": query,
                        "serp_page": page,
                        "error_type": type(exc).__name__,
                        "error": str(exc),
                        "recorded_at": datetime.now(timezone.utc).isoformat(),
                    })
                    query_failed = True
                    break
                finally:
                    if response is not None:
                        response.close()

                if not organic:
                    break
                retrieved_at = datetime.now(timezone.utc).isoformat()
                new_url_count = 0
                for fallback_position, result in enumerate(organic, start=1):
                    candidate_url = normalise_url(result.get("link", ""))
                    if not candidate_url or candidate_url in seen_urls:
                        continue
                    candidate_domain = registered_domain(candidate_url)
                    page_position = int(result.get("position") or fallback_position)
                    rank = (
                        page_position
                        if page_position > SERP_PAGE_SIZE
                        else ((page - 1) * SERP_PAGE_SIZE) + page_position
                    )
                    row = {
                        "snapshot_id": SNAPSHOT_ID,
                        "query_id": query_id,
                        "query": query,
                        "serp_page": page,
                        "page_position": page_position,
                        "rank": rank,
                        "serp_url": result.get("link", ""),
                        "candidate_url": candidate_url,
                        "serp_title": result.get("title", ""),
                        "serp_snippet": result.get("snippet", ""),
                        "retrieved_at": retrieved_at,
                    }
                    seen_urls.add(candidate_url)
                    query_rows.append(row)
                    rows_by_key[(query_id, candidate_url)] = row
                    new_url_count += 1
                    if len(query_rows) >= RESULTS_PER_QUERY:
                        break

                save_query_hits(list(rows_by_key.values()))
                if new_url_count == 0:
                    break
                page += 1
                time.sleep(SERPER_REQUEST_DELAY_SECONDS)

            if not query_failed:
                completed_query_ids.add(query_id)
            write_json(SERP_PROGRESS_FILE, {
                "completed_query_ids": sorted(completed_query_ids),
            })
    finally:
        session.close()

    frame = save_query_hits(list(rows_by_key.values()))
    expected_query_ids = {query_id_for(query) for query in QUERIES}
    write_checkpoint(
        "collection",
        "complete" if expected_query_ids <= completed_query_ids else "partial",
        query_hit_count=len(frame),
        completed_query_count=len(completed_query_ids),
        expected_query_count=len(expected_query_ids),
    )
    return frame
````

## 4.4 Redirect-aware page requests

Each candidate URL is normalised, checked against robots policy, and requested without automatic redirects. Every redirect target is normalised, checked for a public address, and evaluated against its own robots policy before retrieval. YouTube observations are retained as deferred media records and are not mistaken for analysable article text.


````python
REDIRECT_STATUS_CODES = {301, 302, 303, 307, 308}
YOUTUBE_DOMAINS = {"youtube.com", "youtu.be"}

def is_deferred_media_url(url):
    return registered_domain(url) in YOUTUBE_DOMAINS

def deferred_media_record(candidate_url):
    identity_url = normalise_url(candidate_url)
    return {
        "record_type": "page_fetch",
        "snapshot_id": SNAPSHOT_ID,
        "page_id": stable_id("page", identity_url),
        "candidate_url": candidate_url,
        "resolved_identity_url": identity_url,
        "fetch_status": "deferred_media",
        "analysis_status": "transcript_required",
        "failure_reason": "youtube_transcript_workflow_deferred",
        "recorded_at": datetime.now(timezone.utc).isoformat(),
    }

def request_page_with_policy(candidate_url, session):
    current_url = normalise_url(candidate_url)
    if not current_url:
        return {
            "response": None,
            "failed_url": candidate_url,
            "failure_reason": "invalid_candidate_url",
            "error": "Candidate is not a valid HTTP(S) URL",
            "redirect_chain": [],
            "robots_checks": [],
        }

    redirect_chain = []
    robots_checks = []
    visited_urls = set()

    for _ in range(MAX_REDIRECTS + 1):
        if current_url in visited_urls:
            return {
                "response": None,
                "failed_url": current_url,
                "failure_reason": "redirect_loop",
                "error": "Redirect loop detected",
                "redirect_chain": redirect_chain,
                "robots_checks": robots_checks,
            }
        visited_urls.add(current_url)

        try:
            robots = robots_allowed(current_url, session)
        except Exception as exc:
            return {
                "response": None,
                "failed_url": current_url,
                "failure_reason": "robots_check_error",
                "error": f"{type(exc).__name__}: {exc}",
                "redirect_chain": redirect_chain,
                "robots_checks": robots_checks,
            }

        robots_checks.append({"url": current_url, **robots})
        if not robots["allowed"]:
            return {
                "response": None,
                "blocked_url": current_url,
                "failed_url": None,
                "failure_reason": robots["reason"],
                "error": robots.get("error"),
                "redirect_chain": redirect_chain,
                "robots_checks": robots_checks,
            }

        crawl_delay = robots.get("crawl_delay")
        request_delay = max(
            REQUEST_DELAY_SECONDS,
            float(crawl_delay) if crawl_delay is not None else 0.0,
        )
        response = None
        try:
            response = polite_get(
                session, current_url, request_delay,
                stream=True, allow_redirects=False,
            )
            location = response.headers.get("location")
            if response.status_code in REDIRECT_STATUS_CODES and location:
                next_url = redirect_request_url(current_url, location)
                redirect_chain.append({
                    "from": current_url,
                    "status": response.status_code,
                    "to": next_url,
                })
                response.close()
                response = None
                current_url = next_url
                continue
            return {
                "response": response,
                "blocked_url": None,
                "failed_url": None,
                "failure_reason": None,
                "error": None,
                "redirect_chain": redirect_chain,
                "robots_checks": robots_checks,
            }
        except Exception as exc:
            if response is not None:
                response.close()
            return {
                "response": None,
                "failed_url": current_url,
                "failure_reason": "request_error",
                "error": f"{type(exc).__name__}: {exc}",
                "redirect_chain": redirect_chain,
                "robots_checks": robots_checks,
            }

    return {
        "response": None,
        "failed_url": current_url,
        "failure_reason": "too_many_redirects",
        "error": f"More than {MAX_REDIRECTS} redirects",
        "redirect_chain": redirect_chain,
        "robots_checks": robots_checks,
    }
````

## 4.5 BeautifulSoup extraction and evidence construction

BeautifulSoup provides the transparent baseline representation. The extractor stores a private content-addressed copy of permitted HTML, then records the exact content hash, parser strategy, encoding, title, meta description, headings, cleaned body, and bounded evidence fragments. The later critique will test whether this `main -> article -> body` strategy preserves content better than alternative extractors.


### Acquisition expectation: normal scrape loss

The Serper result set and the BeautifulSoup scrape are deliberately treated as two different layers. Serper records the search visibility sample; Requests and BeautifulSoup record the accessible text subset.

A 60-80% successful unique-page fetch rate is a normal operating range for this workflow. Some returned pages will block automated access, restrict robots access, return HTTP errors such as 403 or 429, depend on JavaScript rendering, expose unsupported content types, or stall during extraction. This does not by itself mean the run failed.

Escalate only when the success rate falls materially below that range, failures are concentrated in a way that biases the sample, or the loss is caused by notebook code rather than target-site access conditions.


````python
EXTRACTION_STRATEGY = "beautifulsoup_main_article_body_v1"
EXPECTED_UNIQUE_FETCH_SUCCESS_RANGE = (0.60, 0.80)
SCRAPE_SUCCESS_OPERATING_NOTE = (
    "Requests/BeautifulSoup normally retrieves an accessible text subset rather than every URL returned by Serper; "
    "historical successful unique-page fetch rates commonly fall around 60-80%."
)
SAFE_RESPONSE_HEADERS = {
    "content-type", "content-length", "etag",
    "last-modified", "cache-control",
}

def body_fragments(text, maximum_characters=1200):
    paragraphs = [
        " ".join(part.split())
        for part in re.split(r"\n\s*\n|(?<=[.!?])\s+(?=[A-Z])", text)
        if part.strip()
    ]
    fragments = []
    buffer = ""
    for paragraph in paragraphs:
        candidate = f"{buffer} {paragraph}".strip()
        if buffer and len(candidate) > maximum_characters:
            fragments.append(buffer)
            buffer = paragraph
        else:
            buffer = candidate
    if buffer:
        fragments.append(buffer)
    return fragments

def safe_response_headers(response):
    return {
        key.lower(): value
        for key, value in response.headers.items()
        if key.lower() in SAFE_RESPONSE_HEADERS
    }

def cache_source_bytes(content, content_hash):
    if not RAW_HTML_CACHE_ENABLED:
        return None
    path = assert_private_output_path(RAW_HTML_DIR / f"{content_hash}.html")
    path.parent.mkdir(parents=True, exist_ok=True)
    if not path.exists():
        path.write_bytes(content)
    return path.relative_to(OUTPUT_DIR).as_posix()

def declared_canonical_url(soup, final_url):
    tag = soup.find(
        "link",
        rel=lambda value: value and "canonical" in [
            str(item).lower()
            for item in (value if isinstance(value, list) else [value])
        ],
    )
    if not tag or not tag.get("href"):
        return ""
    return normalise_url(urljoin(final_url, tag.get("href", "")))

def extract_page(candidate_url, session):
    if is_deferred_media_url(candidate_url):
        return deferred_media_record(candidate_url)

    request_result = request_page_with_policy(candidate_url, session)
    robots_checks = request_result.get("robots_checks", [])
    redirect_chain = request_result.get("redirect_chain", [])
    response = request_result.get("response")

    if response is None:
        status = (
            "robots_blocked"
            if request_result.get("blocked_url")
            else "fetch_error"
        )
        return {
            "record_type": "page_fetch",
            "snapshot_id": SNAPSHOT_ID,
            "candidate_url": candidate_url,
            "final_fetch_url": request_result.get("blocked_url") or request_result.get("failed_url"),
            "fetch_status": status,
            "failure_reason": request_result.get("failure_reason"),
            "error": request_result.get("error"),
            "redirect_chain": redirect_chain,
            "robots_checks": robots_checks,
            "recorded_at": datetime.now(timezone.utc).isoformat(),
        }

    try:
        final_url = normalise_url(response.url)
        base_record = {
            "record_type": "page_fetch",
            "snapshot_id": SNAPSHOT_ID,
            "candidate_url": candidate_url,
            "final_fetch_url": final_url,
            "http_status": response.status_code,
            "response_headers": safe_response_headers(response),
            "redirect_chain": redirect_chain,
            "robots_checks": robots_checks,
        }

        if response.status_code in {401, 403, 451}:
            return {
                **base_record,
                "fetch_status": "server_refused",
                "failure_reason": "explicit_http_refusal",
            }
        retry_after = response.headers.get("Retry-After")
        if response.status_code == 429 or (response.status_code == 503 and retry_after):
            return {
                **base_record,
                "fetch_status": "rate_limited",
                "failure_reason": "explicit_retry_later",
                "retry_after": retry_after,
            }
        response.raise_for_status()

        content_type = response.headers.get("content-type", "").lower()
        if not any(value in content_type for value in {"text/html", "application/xhtml+xml"}):
            return {
                **base_record,
                "fetch_status": "unsupported_content_type",
                "failure_reason": "not_html",
            }

        source_bytes = read_limited_response_bytes(response, MAX_RESPONSE_BYTES)
        content_hash = sha256_bytes(source_bytes)
        soup = BeautifulSoup(source_bytes, "lxml")
        declared_canonical = declared_canonical_url(soup, final_url)
        if declared_canonical and registered_domain(declared_canonical) == registered_domain(final_url):
            identity_url = declared_canonical
            canonical_decision = "same_domain_declared_canonical"
        elif declared_canonical:
            identity_url = final_url
            canonical_decision = "cross_domain_canonical_not_trusted"
        else:
            identity_url = final_url
            canonical_decision = "final_fetch_url"

        title = soup.title.get_text(" ", strip=True) if soup.title else ""
        meta_tag = soup.find("meta", attrs={"name": re.compile(r"^description$", re.I)})
        meta_description = meta_tag.get("content", "").strip() if meta_tag else ""
        main = soup.find("main") or soup.find("article") or soup.body or soup
        for element in main.find_all([
            "nav", "footer", "aside", "form", "script",
            "style", "noscript", "svg", "template",
        ]):
            element.decompose()
        headings = [
            {"level": heading.name.lower(), "text": heading.get_text(" ", strip=True)}
            for heading in main.find_all(re.compile(r"^h[1-6]$"))
            if heading.get_text(" ", strip=True)
        ]
        clean_body = main.get_text("\n", strip=True)

        evidence = []
        if title:
            evidence.append({"evidence_id": "ev_title_01", "field": "title", "text": title})
        if meta_description:
            evidence.append({
                "evidence_id": "ev_meta_01",
                "field": "meta_description",
                "text": meta_description,
            })
        heading_counts = Counter()
        for heading in headings:
            heading_counts[heading["level"]] += 1
            evidence.append({
                "evidence_id": f"ev_{heading['level']}_{heading_counts[heading['level']]:02d}",
                "field": heading["level"],
                "text": heading["text"],
            })
        for index, text in enumerate(body_fragments(clean_body), start=1):
            evidence.append({
                "evidence_id": f"ev_body_{index:03d}",
                "field": "body",
                "text": text,
            })

        return {
            **base_record,
            "page_id": stable_id("page", identity_url),
            "declared_canonical_url": declared_canonical,
            "resolved_identity_url": identity_url,
            "canonical_decision": canonical_decision,
            "fetch_status": "success",
            "analysis_status": "html_evidence_ready",
            "fetched_at": datetime.now(timezone.utc).isoformat(),
            "content_sha256": content_hash,
            "content_bytes": len(source_bytes),
            "source_encoding": soup.original_encoding,
            "private_source_path": cache_source_bytes(source_bytes, content_hash),
            "extraction_strategy": EXTRACTION_STRATEGY,
            "title": title,
            "meta_description": meta_description,
            "h1": next((item["text"] for item in headings if item["level"] == "h1"), ""),
            "headings": headings,
            "clean_body": clean_body,
            "evidence": evidence,
        }
    except Exception as exc:
        return {
            "record_type": "page_fetch",
            "snapshot_id": SNAPSHOT_ID,
            "candidate_url": candidate_url,
            "final_fetch_url": normalise_url(response.url),
            "fetch_status": "fetch_error",
            "failure_reason": "request_or_parse_error",
            "error": f"{type(exc).__name__}: {exc}",
            "redirect_chain": redirect_chain,
            "robots_checks": robots_checks,
            "recorded_at": datetime.now(timezone.utc).isoformat(),
        }
    finally:
        response.close()
````

## 4.6 Snapshot orchestration and manifest

The orchestration stage deduplicates candidate URLs, resumes terminal fetch records, collapses same-identity page variants, preserves aliases, maps queries back to resolved page identities, and writes a manifest that fingerprints both the notebook source and acquisition configuration. All artifacts remain private.


````python
def fetch_pages(query_hits):
    if ACQUISITION_ACTION not in {"acquire", "resume"}:
        raise RuntimeError("Page fetching is available only in acquire or resume mode.")
    if not NETWORK_APPROVED:
        raise PermissionError("Human approval is required before network acquisition.")

    session = make_session()
    existing_records = read_jsonl(PAGE_FETCHES_FILE)
    records_by_url = {
        row["candidate_url"]: row
        for row in existing_records
        if row.get("candidate_url") and row.get("fetch_status")
    }
    target_urls = sorted(set(query_hits["candidate_url"].dropna()))

    try:
        for url in tqdm(target_urls, desc="Pages"):
            if url in records_by_url:
                continue
            record = extract_page(url, session)
            records_by_url[url] = record
            append_jsonl(PAGE_FETCHES_FILE, record)
    finally:
        session.close()

    records = [records_by_url[url] for url in target_urls if url in records_by_url]
    write_jsonl(PAGE_FETCHES_FILE, records)
    successful = [row for row in records if row.get("fetch_status") == "success"]
    deferred = [row for row in records if row.get("fetch_status") == "deferred_media"]
    failures = [
        row for row in records
        if row.get("fetch_status") not in {"success", "deferred_media"}
    ]

    grouped = defaultdict(list)
    candidate_to_page = {}
    for row in records:
        if row.get("page_id"):
            candidate_to_page[row["candidate_url"]] = row["page_id"]
    for row in successful:
        grouped[row["page_id"]].append(row)

    pages = []
    for page_id, variants in grouped.items():
        variants.sort(
            key=lambda row: (len(row.get("evidence", [])), len(row.get("clean_body", ""))),
            reverse=True,
        )
        page = dict(variants[0])
        page["identity_aliases"] = sorted({row["candidate_url"] for row in variants})
        pages.append(page)
    pages.sort(key=lambda row: row["page_id"])
    write_jsonl(PAGES_EXTRACTED_FILE, pages)

    query_hits = query_hits.copy()
    query_hits["page_id"] = query_hits["candidate_url"].map(candidate_to_page)
    output = assert_private_output_path(QUERY_HITS_FILE)
    query_hits.to_csv(output, index=False)
    write_checkpoint(
        "fetch",
        "complete" if len(records) == len(target_urls) else "partial",
        target_url_count=len(target_urls),
        fetch_record_count=len(records),
        successful_page_count=len(pages),
        deferred_media_count=len(deferred),
        failure_count=len(failures),
    )
    return pages, records, pd.DataFrame(failures), pd.DataFrame(deferred), query_hits

def build_acquisition_manifest(query_hits, pages, records):
    return {
        "run_name": RUN_NAME,
        "snapshot_id": SNAPSHOT_ID,
        "started_at": RUN_STARTED_AT,
        "completed_at": datetime.now(timezone.utc).isoformat(),
        "notebook_source_sha256": notebook_source_hash(),
        "package_versions": package_versions(),
        "queries": QUERIES,
        "search_configuration": {
            "provider": "Serper Google Search API",
            "country": GOOGLE_COUNTRY,
            "language": GOOGLE_LANGUAGE,
            "results_per_query": RESULTS_PER_QUERY,
            "primary_domain_cap": PRIMARY_DOMAIN_CAP,
            "domain_cap_sensitivity": DOMAIN_CAP_SENSITIVITY,
            "domain_capping_note": "Primary acquisition preserves top n results as returned; caps are post-hoc sensitivity views.",
            "sampling_protocol_version": SAMPLING_PROTOCOL_VERSION,
            "query_plan": QUERY_PLAN,
            "serp_page_size": SERP_PAGE_SIZE,
        },
        "request_configuration": {
            "user_agent": USER_AGENT,
            "timeout_seconds": REQUEST_TIMEOUT_SECONDS,
            "minimum_domain_delay_seconds": REQUEST_DELAY_SECONDS,
            "maximum_redirects": MAX_REDIRECTS,
            "maximum_response_bytes": MAX_RESPONSE_BYTES,
            "robots_unavailable_policy": ROBOTS_UNAVAILABLE_POLICY,
        },
        "extraction_configuration": {
            "strategy": EXTRACTION_STRATEGY,
            "parser": "lxml",
            "private_raw_html_cache": RAW_HTML_CACHE_ENABLED,
            "youtube_policy": "record_as_transcript_required",
        },
        "counts": {
            "query_hits": len(query_hits),
            "fetch_records": len(records),
            "successful_unique_pages": len(pages),
            "deferred_media": sum(row.get("fetch_status") == "deferred_media" for row in records),
            "terminal_failures": sum(
                row.get("fetch_status") not in {"success", "deferred_media"}
                for row in records
            ),
        },
        "private_output_files": [
            QUERY_HITS_FILE.name, SERP_ERRORS_FILE.name, PAGE_FETCHES_FILE.name,
            PAGES_EXTRACTED_FILE.name, MANIFEST_FILE.name,
        ],
    }

def run_acquisition():
    if ACQUISITION_ACTION not in {"acquire", "resume"}:
        raise RuntimeError("Set ACQUISITION_ACTION to acquire or resume.")
    if not NETWORK_APPROVED:
        raise PermissionError("Human approval is required before the first network run.")
    if SNAPSHOT_ID == "not-created":
        raise RuntimeError("A timestamped snapshot ID was not created. Rerun configuration.")

    assert_private_output_path(OUTPUT_DIR).mkdir(parents=True, exist_ok=True)
    query_hits = collect_serp_results()
    pages, records, failures, deferred, query_hits = fetch_pages(query_hits)
    manifest = build_acquisition_manifest(query_hits, pages, records)
    write_json(MANIFEST_FILE, manifest)
    write_checkpoint(
        "acquisition", "complete",
        manifest=MANIFEST_FILE.name,
        successful_page_count=len(pages),
    )
    return {
        "query_hits": query_hits,
        "pages": pages,
        "fetch_records": records,
        "failures": failures,
        "deferred_media": deferred,
        "manifest": manifest,
    }

def load_acquisition_snapshot(snapshot_id):
    snapshot_dir = LOCAL_DATA_ROOT / RUN_NAME / snapshot_id
    assert_private_output_path(snapshot_dir)
    query_hits_path = snapshot_dir / "query_hits.csv"
    if not query_hits_path.exists():
        raise FileNotFoundError(f"Snapshot does not contain query hits: {snapshot_id}")
    records = read_jsonl(snapshot_dir / "page_fetches.jsonl")
    failures = [row for row in records if row.get("fetch_status") not in {"success", "deferred_media"}]
    deferred = [row for row in records if row.get("fetch_status") == "deferred_media"]
    return {
        "query_hits": pd.read_csv(query_hits_path),
        "pages": read_jsonl(snapshot_dir / "pages_extracted.jsonl"),
        "fetch_records": records,
        "failures": pd.DataFrame(failures),
        "deferred_media": pd.DataFrame(deferred),
        "manifest": read_json(snapshot_dir / "run_manifest.json", {}),
    }
````

# Stage 5 - Controlled acquisition pilot

Stage 5 runs the Stage 4 collector against a deliberately small SERP sample before any full acquisition. The pilot audits URL coverage, robots outcomes, extraction success, empty fields, duplicate source bodies, and private-path containment. The committed notebook remains inert: a private run must explicitly set the acquisition environment variables.


````python
import numpy as np
def acquisition_quality_report(acquisition_result):
    if acquisition_result is None:
        return {"status": "not_run"}
    records = acquisition_result["fetch_records"]
    pages = acquisition_result["pages"]
    statuses = Counter(row.get("fetch_status", "unknown") for row in records)
    body_hashes = Counter(
        row.get("content_sha256") for row in pages if row.get("content_sha256")
    )
    body_lengths = [len(row.get("clean_body", "")) for row in pages]
    report = {
        "status": "complete",
        "snapshot_id": SNAPSHOT_ID,
        "requested_results": RESULTS_PER_QUERY * len(QUERIES),
        "query_hits": len(acquisition_result["query_hits"]),
        "fetch_records": len(records),
        "unique_pages": len(pages),
        "fetch_status_counts": dict(sorted(statuses.items())),
        "success_rate": round(statuses.get("success", 0) / max(len(records), 1), 4),
        "empty_title_pages": sum(not row.get("title", "").strip() for row in pages),
        "empty_body_pages": sum(not row.get("clean_body", "").strip() for row in pages),
        "short_body_pages_under_500_chars": sum(length < 500 for length in body_lengths),
        "median_body_characters": float(pd.Series(body_lengths).median()) if body_lengths else 0.0,
        "duplicate_content_hash_groups": sum(count > 1 for count in body_hashes.values()),
        "private_output_contained": Path(OUTPUT_DIR).resolve().is_relative_to(LOCAL_DATA_ROOT.resolve()),
    }
    write_json(OUTPUT_DIR / "stage5_acquisition_quality.json", report)
    write_checkpoint("stage5_pilot_audit", "complete", **{key: value for key, value in report.items() if key != "status"})
    return report


def domain_concentration_report(acquisition_result):
    if acquisition_result is None:
        return pd.DataFrame(), {"status": "not_run"}
    hits = acquisition_result["query_hits"].copy()
    if hits.empty:
        return pd.DataFrame(), {"status": "empty"}
    hits["domain"] = hits["candidate_url"].map(registered_domain)
    rows = []
    for query, group in hits.groupby("query"):
        counts = group["domain"].value_counts()
        probabilities = counts / max(counts.sum(), 1)
        entropy = float(-(probabilities * np.log2(probabilities)).sum()) if len(probabilities) else 0.0
        rows.append({
            "query": query,
            "results": int(len(group)),
            "unique_domains": int(counts.size),
            "top_domain": counts.index[0] if len(counts) else "",
            "top_domain_results": int(counts.iloc[0]) if len(counts) else 0,
            "top_domain_share": float(counts.iloc[0] / len(group)) if len(group) else 0.0,
            "duplicate_domain_results": int(len(group) - counts.size),
            "duplicate_domain_share": float((len(group) - counts.size) / len(group)) if len(group) else 0.0,
            "domain_entropy": entropy,
        })
    report = pd.DataFrame(rows).sort_values(["top_domain_share", "duplicate_domain_share"], ascending=False)
    summary = {
        "status": "complete",
        "queries": int(report["query"].nunique()) if not report.empty else 0,
        "total_results": int(len(hits)),
        "unique_domains": int(hits["domain"].nunique()),
        "mean_top_domain_share": float(report["top_domain_share"].mean()) if not report.empty else None,
        "mean_duplicate_domain_share": float(report["duplicate_domain_share"].mean()) if not report.empty else None,
    }
    report.to_csv(assert_private_output_path(OUTPUT_DIR / "domain_concentration_by_query.csv"), index=False)
    write_json(OUTPUT_DIR / "domain_concentration_summary.json", summary)
    return report, summary


stage5_quality = acquisition_quality_report(acquisition)
if acquisition is not None:
    domain_concentration, domain_concentration_summary = domain_concentration_report(acquisition)
    display(pd.DataFrame([stage5_quality | {f'domain_{key}': value for key, value in domain_concentration_summary.items() if key != 'status'}]))
    display(domain_concentration.head(20))
else:
    domain_concentration, domain_concentration_summary = pd.DataFrame(), {'status': 'not_run'}
    print("Stage 5 prepared. Set the documented environment gates for a private pilot run.")
````

````python
def page_analysis_text(page):
    headings = ' '.join(item.get('text', '') for item in page.get('headings', []))
    fields = [page.get('title', ''), page.get('meta_description', ''), headings, page.get('clean_body', '')]
    return ' '.join(value for value in fields if value).strip()[:MAX_ANALYSIS_CHARACTERS]

def prepare_analysis_documents(pages):
    import spacy
    nlp = spacy.blank('en')
    rows = []
    for page in pages:
        text = page_analysis_text(page)
        doc = nlp.make_doc(text)
        tokens = [token.lower_ for token in doc if token.is_alpha and len(token.text) > 2 and not token.is_stop]
        if tokens:
            rows.append({'page_id': page['page_id'], 'url': page.get('resolved_identity_url') or page.get('final_fetch_url'), 'domain': registered_domain(page.get('resolved_identity_url') or page.get('final_fetch_url', '')), 'raw_text': text, 'token_text': ' '.join(tokens)})
    return pd.DataFrame(rows)

def fit_tfidf(documents):
    from sklearn.feature_extraction.text import TfidfVectorizer
    minimum_document_frequency = 2 if len(documents) >= 10 else 1
    maximum_document_frequency = 0.95 if len(documents) >= 20 else 1.0
    vectorizer = TfidfVectorizer(ngram_range=(1, 3), min_df=minimum_document_frequency, max_df=maximum_document_frequency, max_features=MAX_TFIDF_FEATURES, sublinear_tf=True)
    return vectorizer, vectorizer.fit_transform(documents['token_text'])

def raw_phrase_candidate_table(documents, vectorizer, matrix):
    names = vectorizer.get_feature_names_out()
    rows = []
    for index, page in documents.reset_index(drop=True).iterrows():
        vector = matrix.getrow(index)
        ranked = sorted(zip(vector.indices, vector.data), key=lambda item: item[1], reverse=True)
        for rank, (term_index, score) in enumerate(ranked[:TOP_CONCEPTS_PER_PAGE], start=1):
            rows.append({'page_id': page['page_id'], 'url': page['url'], 'domain': page['domain'], 'raw_surface': names[term_index], 'tfidf': float(score), 'page_concept_rank': rank, 'representation_role': 'raw_tfidf_candidate'})
    return pd.DataFrame(rows)

def normalise_concept_surface(value):
    return ' '.join(token for token in re.findall(r'[a-z0-9]+', str(value).lower()) if len(token) > 1)

def concept_quality_flags(surface):
    generic = {'example', 'examples', 'questions', 'question', 'topic', 'topics', 'page', 'pages', 'post', 'posts', 'weeks', 'wrote', 'existing', 'find', 'relevant', 'topically', 'optimized', 'techniques', 'implementing', 'competitor', 'paleo', 'diet'}
    tokens = surface.split()
    flags = []
    if len(tokens) == 1 and tokens[0] in generic:
        flags.append('generic_single_word')
    if len(tokens) > 3:
        flags.append('long_ngram')
    if not tokens:
        flags.append('empty_after_normalisation')
    return '|'.join(flags)


REVIEW_TARGET_TERMS = {
    'semantic seo', 'python', 'python seo', 'python for seo', 'seo automation', 'nlp', 'spacy', 'bert', 'roberta',
    'semantic similarity', 'keyword clustering', 'networkx', 'network analysis', 'machine learning', 'content gap',
    'entity', 'entities', 'embedding', 'embeddings', 'semantic', 'search', 'internal link', 'internal linking',
}
REVIEW_ARTIFACT_TERMS = {
    'ahrefs', 'diwakar', 'loomba', 'july', 'ryan', 'replit', 'paleo', 'diet', 'mobile', 'wrote', 'post', 'page',
    'example', 'questions', 'general', 'existing', 'find', 'need', 'weeks', 'topically', 'optimized', 'relevant',
}


def review_target_matches(concept):
    concept = normalise_concept_surface(concept)
    matches = sorted(term for term in REVIEW_TARGET_TERMS if term in concept or concept in term)
    return ' | '.join(matches)


def likely_source_or_page_artifact(concept):
    tokens = set(normalise_concept_surface(concept).split())
    return bool(tokens and tokens & REVIEW_ARTIFACT_TERMS)


def suggested_review_action(row):
    flags = str(row.get('automatic_flags', ''))
    concept = str(row.get('proposed_canonical', ''))
    page_support = int(row.get('page_support', 0) or 0)
    domain_support = int(row.get('domain_support', 0) or 0)
    target_match = str(row.get('target_term_match', ''))
    existing_decision = str(row.get('decision', '')).strip().lower()
    if existing_decision in {'accept', 'rename', 'reject'}:
        return existing_decision, 'Existing human or prior review decision is being preserved.'
    if likely_source_or_page_artifact(concept):
        return 'reject_likely', 'Looks like a source, author, date, page, or generic SERP artefact rather than a reusable Semantic SEO concept.'
    if page_support >= 2 and domain_support >= 2 and target_match:
        return 'inspect_for_accept_or_rename', 'Recurring across independent domains and aligned with the target Semantic SEO/Python vocabulary.'
    if page_support >= 2 and domain_support >= 2:
        return 'inspect_for_possible_concept', 'Recurring across independent domains, but not clearly tied to the target vocabulary.'
    if 'generic_single_word' in flags:
        return 'reject_likely', 'Single-word generic candidate with weak evidence; review only if the graph context suggests strategic relevance.'
    return 'defer_low_priority', 'Single-page or weak-support candidate; keep in the audit trail but do not review before higher-priority batches.'


def review_batch_for_action(action, priority):
    if action == 'inspect_for_accept_or_rename':
        return 'batch_1_high_priority_target_aligned'
    if action == 'inspect_for_possible_concept':
        return 'batch_2_possible_concepts'
    if action == 'reject_likely':
        return 'batch_3_likely_rejects'
    if priority >= 4:
        return 'batch_2_possible_concepts'
    return 'batch_4_low_priority_audit_only'


def enrich_concept_review_ledger(ledger, raw_nodes):
    if ledger.empty:
        return ledger
    enriched = ledger.copy()
    enriched['target_term_match'] = enriched['proposed_canonical'].map(review_target_matches)
    enriched['likely_artifact'] = enriched['proposed_canonical'].map(likely_source_or_page_artifact)
    if not raw_nodes.empty and 'concept' in raw_nodes.columns:
        node_context = raw_nodes.rename(columns={'concept': 'proposed_canonical', 'community': 'raw_community', 'pagerank': 'raw_pagerank', 'weighted_degree': 'raw_weighted_degree'})
        context_columns = [column for column in ['proposed_canonical', 'raw_community', 'raw_pagerank', 'raw_weighted_degree'] if column in node_context.columns]
        enriched = enriched.merge(node_context[context_columns], on='proposed_canonical', how='left')
    for column, default in [('raw_community', -1), ('raw_pagerank', 0.0), ('raw_weighted_degree', 0.0)]:
        if column not in enriched.columns:
            enriched[column] = default
        enriched[column] = enriched[column].fillna(default)
    enriched['review_priority'] = (
        enriched['page_support'].astype(float) * 3.0
        + enriched['domain_support'].astype(float) * 4.0
        + enriched['raw_pagerank'].astype(float) * 25.0
        + enriched['raw_weighted_degree'].astype(float) * 1.5
        + enriched['target_term_match'].ne('').astype(float) * 5.0
        - enriched['automatic_flags'].astype(str).str.contains('generic_single_word', regex=False).astype(float) * 2.0
        - enriched['likely_artifact'].astype(float) * 5.0
    ).round(3)
    suggestions = enriched.apply(suggested_review_action, axis=1, result_type='expand')
    enriched['suggested_action'] = suggestions[0]
    enriched['suggested_reason'] = suggestions[1]
    enriched['review_batch'] = [review_batch_for_action(action, priority) for action, priority in zip(enriched['suggested_action'], enriched['review_priority'])]
    ordered_columns = [
        'candidate_id', 'review_batch', 'review_priority', 'suggested_action', 'suggested_reason', 'proposed_canonical',
        'raw_surfaces', 'target_term_match', 'page_support', 'domain_support', 'raw_community', 'raw_pagerank',
        'raw_weighted_degree', 'mean_tfidf', 'automatic_flags', 'likely_artifact', 'decision', 'human_canonical',
        'human_note', 'decision_authority', 'review_required'
    ]
    return enriched[[column for column in ordered_columns if column in enriched.columns]].sort_values(['review_batch', 'review_priority', 'proposed_canonical'], ascending=[True, False, True]).reset_index(drop=True)


def build_review_batch_summary(review_ledger):
    columns = ['review_batch', 'candidate_count', 'top_candidates', 'suggested_actions', 'mean_priority', 'next_human_action']
    if review_ledger.empty:
        return pd.DataFrame(columns=columns)
    rows = []
    guidance = {
        'batch_1_high_priority_target_aligned': 'Review first: accept, rename, merge, or reject these before scaling the sample.',
        'batch_2_possible_concepts': 'Review after Batch 1 if they appear strategically meaningful.',
        'batch_3_likely_rejects': 'Spot-check; reject in bulk only after confirming the pattern.',
        'batch_4_low_priority_audit_only': 'Leave collapsed unless a later graph or article question makes them relevant.',
    }
    for batch, group in review_ledger.groupby('review_batch', sort=True):
        ordered = group.sort_values('review_priority', ascending=False)
        rows.append({
            'review_batch': batch,
            'candidate_count': int(len(group)),
            'top_candidates': ' | '.join(ordered['proposed_canonical'].astype(str).head(8)),
            'suggested_actions': ' | '.join(sorted(set(ordered['suggested_action'].astype(str)))),
            'mean_priority': round(float(ordered['review_priority'].mean()), 3),
            'next_human_action': guidance.get(batch, 'Inspect only if needed.'),
        })
    return pd.DataFrame(rows, columns=columns)


def build_community_review_summary(review_ledger):
    columns = ['raw_community', 'candidate_count', 'review_first', 'target_terms', 'likely_rejects', 'provisional_contents', 'human_question']
    if review_ledger.empty or 'raw_community' not in review_ledger.columns:
        return pd.DataFrame(columns=columns)
    rows = []
    for community, group in review_ledger.groupby('raw_community', dropna=False):
        ordered = group.sort_values('review_priority', ascending=False)
        target_terms = ordered[ordered['target_term_match'].astype(str).ne('')]['proposed_canonical'].astype(str).head(6).tolist()
        likely_rejects = ordered[ordered['suggested_action'].eq('reject_likely')]['proposed_canonical'].astype(str).head(6).tolist()
        review_first = ordered['proposed_canonical'].astype(str).head(8).tolist()
        rows.append({
            'raw_community': int(community) if pd.notna(community) else -1,
            'candidate_count': int(len(group)),
            'review_first': ' | '.join(review_first),
            'target_terms': ' | '.join(target_terms),
            'likely_rejects': ' | '.join(likely_rejects),
            'provisional_contents': ' / '.join(review_first[:4]),
            'human_question': 'Does this blob contain a reusable Semantic SEO concept, or mostly page/source artefacts?',
        })
    return pd.DataFrame(rows, columns=columns).sort_values(['raw_community']).reset_index(drop=True)

def build_concept_review_ledger(raw_candidates):
    columns = ['candidate_id', 'proposed_canonical', 'raw_surfaces', 'page_support', 'domain_support', 'mean_tfidf', 'automatic_flags', 'decision', 'human_canonical', 'human_note', 'decision_authority', 'review_required']
    if raw_candidates.empty:
        return pd.DataFrame(columns=columns)
    working = raw_candidates.copy()
    working['proposed_canonical'] = working['raw_surface'].map(normalise_concept_surface)
    working = working[working['proposed_canonical'].ne('')].copy()
    grouped = working.groupby('proposed_canonical', as_index=False).agg(raw_surfaces=('raw_surface', lambda values: ' | '.join(sorted(set(values)))), page_support=('page_id', 'nunique'), domain_support=('domain', 'nunique'), mean_tfidf=('tfidf', 'mean'))
    grouped['candidate_id'] = grouped['proposed_canonical'].map(lambda value: stable_id('concept', value))
    grouped['automatic_flags'] = grouped['proposed_canonical'].map(concept_quality_flags)
    grouped['decision'] = 'pending'
    grouped['human_canonical'] = ''
    grouped['human_note'] = ''
    grouped['decision_authority'] = 'awaiting_human_review'
    grouped['review_required'] = True
    if CONCEPT_REVIEW_LEDGER_FILE.exists():
        existing = pd.read_csv(CONCEPT_REVIEW_LEDGER_FILE).fillna('').drop_duplicates('candidate_id', keep='last').set_index('candidate_id')
        for field in ['decision', 'human_canonical', 'human_note', 'decision_authority']:
            if field in existing.columns:
                grouped[field] = grouped['candidate_id'].map(existing[field]).fillna(grouped[field])
    grouped['decision'] = grouped['decision'].astype(str).str.strip().str.lower()
    grouped.loc[~grouped['decision'].isin({'pending', 'accept', 'reject', 'rename'}), 'decision'] = 'pending'
    grouped['human_canonical'] = grouped['human_canonical'].astype(str).map(normalise_concept_surface)
    grouped.loc[grouped['decision'].eq('rename') & grouped['human_canonical'].eq(''), 'decision'] = 'pending'
    return grouped[columns].sort_values(['page_support', 'domain_support', 'proposed_canonical'], ascending=[False, False, True]).reset_index(drop=True)

def concept_provenance_table(raw_candidates, ledger):
    columns = ['page_id', 'url', 'domain', 'raw_surface', 'proposed_canonical', 'candidate_id', 'tfidf', 'page_concept_rank', 'decision', 'human_canonical', 'canonical_concept', 'decision_authority', 'automatic_flags']
    if raw_candidates.empty or ledger.empty:
        return pd.DataFrame(columns=columns)
    provenance = raw_candidates.copy()
    provenance['proposed_canonical'] = provenance['raw_surface'].map(normalise_concept_surface)
    provenance['candidate_id'] = provenance['proposed_canonical'].map(lambda value: stable_id('concept', value))
    provenance = provenance.merge(ledger[['candidate_id', 'decision', 'human_canonical', 'decision_authority', 'automatic_flags']], on='candidate_id', how='left', validate='many_to_one')
    provenance['canonical_concept'] = ''
    provenance.loc[provenance['decision'].eq('accept'), 'canonical_concept'] = provenance['proposed_canonical']
    provenance.loc[provenance['decision'].eq('rename'), 'canonical_concept'] = provenance['human_canonical']
    if ALLOW_UNREVIEWED_CONCEPTS:
        provenance.loc[provenance['decision'].eq('pending'), 'canonical_concept'] = provenance['proposed_canonical']
    provenance['canonical_concept'] = provenance['canonical_concept'].map(normalise_concept_surface)
    return provenance[columns].sort_values(['page_id', 'page_concept_rank']).reset_index(drop=True)

def build_raw_phrase_clique_graph(raw_candidates):
    import itertools, math, networkx as nx
    graph = nx.Graph(method='raw_page_wide_tfidf_clique_diagnostic')
    for page_id, group in raw_candidates.groupby('page_id'):
        selected = group.sort_values('page_concept_rank')['raw_surface'].drop_duplicates().tolist()
        domain = group['domain'].iloc[0]
        for concept in selected:
            graph.add_node(concept, pages=set(), domains=set(), node_type='raw_phrase')
            graph.nodes[concept]['pages'].add(page_id)
            graph.nodes[concept]['domains'].add(domain)
        denominator = math.comb(len(selected), 2) if len(selected) > 1 else 1
        for source, target in itertools.combinations(selected, 2):
            if not graph.has_edge(source, target):
                graph.add_edge(source, target, weight=0.0, pages=set(), domains=set(), evidence_windows=0)
            graph[source][target]['weight'] += 1.0 / denominator
            graph[source][target]['pages'].add(page_id)
            graph[source][target]['domains'].add(domain)
    return graph

def build_page_concept_bipartite_graph(provenance):
    import networkx as nx
    graph = nx.Graph(method='reviewed_page_concept_bipartite')
    for row in provenance[provenance['canonical_concept'].ne('')].itertuples(index=False):
        page_node, concept_node = f'page::{row.page_id}', f'concept::{row.canonical_concept}'
        graph.add_node(page_node, node_type='page', page_id=row.page_id, label=row.domain)
        graph.add_node(concept_node, node_type='concept', concept=row.canonical_concept)
        graph.add_edge(page_node, concept_node, weight=float(row.tfidf))
    return graph

def relationship_candidate_table(documents, provenance):
    import itertools
    columns = ['source', 'target', 'page_support', 'domain_support', 'evidence_window_count', 'relationship_type', 'passes_substantive_threshold']
    accepted = provenance[provenance['canonical_concept'].ne('')].copy()
    if accepted.empty:
        return pd.DataFrame(columns=columns)
    terms_by_page = accepted.groupby('page_id')['canonical_concept'].apply(lambda values: sorted(set(values), key=lambda item: (-len(item), item))).to_dict()
    document_by_page = documents.set_index('page_id')
    evidence = {}
    for page_id, terms in terms_by_page.items():
        if page_id not in document_by_page.index:
            continue
        document = document_by_page.loc[page_id]
        segments = [segment.strip().lower() for segment in re.split(r'(?<=[.!?])\s+|\n+', document['raw_text']) if len(segment.strip()) >= 40]
        for segment_index, segment in enumerate(segments):
            present = [term for term in terms if re.search(rf'(?<!\w){re.escape(term)}(?!\w)', segment)]
            for source, target in itertools.combinations(sorted(set(present)), 2):
                record = evidence.setdefault((source, target), {'pages': set(), 'domains': set(), 'windows': set()})
                record['pages'].add(page_id)
                record['domains'].add(document['domain'])
                record['windows'].add((page_id, segment_index))
    rows = []
    for (source, target), record in sorted(evidence.items()):
        pages, domains = len(record['pages']), len(record['domains'])
        rows.append({'source': source, 'target': target, 'page_support': pages, 'domain_support': domains, 'evidence_window_count': len(record['windows']), 'relationship_type': 'local_text_co_occurrence_candidate', 'passes_substantive_threshold': pages >= MIN_SUBSTANTIVE_PAGE_SUPPORT and domains >= MIN_SUBSTANTIVE_DOMAIN_SUPPORT})
    return pd.DataFrame(rows, columns=columns)

def build_substantive_relationship_graph(provenance, relationship_candidates):
    import networkx as nx
    graph = nx.Graph(method='reviewed_local_cooccurrence_with_independent_recurrence')
    accepted = provenance[provenance['canonical_concept'].ne('')].copy()
    support = accepted.groupby('canonical_concept').agg(page_support=('page_id', 'nunique'), domain_support=('domain', 'nunique')).to_dict('index') if not accepted.empty else {}
    for row in relationship_candidates[relationship_candidates['passes_substantive_threshold']].itertuples(index=False) if not relationship_candidates.empty else []:
        for concept in [row.source, row.target]:
            graph.add_node(concept, node_type='reviewed_concept', **support.get(concept, {}))
        graph.add_edge(row.source, row.target, weight=float(row.domain_support), page_support=int(row.page_support), domain_support=int(row.domain_support), evidence_window_count=int(row.evidence_window_count), relationship_type=row.relationship_type)
    return graph

def graph_tables(graph, method):
    import networkx as nx
    import community as community_louvain
    node_columns = ['concept', 'page_support', 'domain_support', 'weighted_degree', 'pagerank', 'betweenness', 'community', 'method']
    edge_columns = ['source', 'target', 'weight', 'page_support', 'domain_support', 'evidence_window_count', 'relationship_type', 'method']
    if graph.number_of_nodes() == 0:
        return pd.DataFrame(columns=node_columns), pd.DataFrame(columns=edge_columns), {}
    if graph.number_of_edges():
        partition, pagerank, betweenness = community_louvain.best_partition(graph, weight='weight', random_state=CLUSTER_RANDOM_STATE), nx.pagerank(graph, weight='weight'), nx.betweenness_centrality(graph, weight=None)
    else:
        partition = {node: -1 for node in graph.nodes()}
        pagerank = {node: 1 / graph.number_of_nodes() for node in graph.nodes()}
        betweenness = {node: 0.0 for node in graph.nodes()}
    nodes = [{'concept': node, 'page_support': int(data.get('page_support', len(data.get('pages', set())))), 'domain_support': int(data.get('domain_support', len(data.get('domains', set())))), 'weighted_degree': float(graph.degree(node, weight='weight')), 'pagerank': float(pagerank[node]), 'betweenness': float(betweenness[node]), 'community': int(partition[node]), 'method': method} for node, data in graph.nodes(data=True)]
    edges = [{'source': source, 'target': target, 'weight': float(data.get('weight', 0.0)), 'page_support': int(data.get('page_support', len(data.get('pages', set())))), 'domain_support': int(data.get('domain_support', len(data.get('domains', set())))), 'evidence_window_count': int(data.get('evidence_window_count', 0)), 'relationship_type': data.get('relationship_type', 'page_wide_raw_clique'), 'method': method} for source, target, data in graph.edges(data=True)]
    return pd.DataFrame(nodes, columns=node_columns), pd.DataFrame(edges, columns=edge_columns), partition

def write_graphml_safely(graph, path):
    import networkx as nx
    safe = nx.Graph(method=graph.graph.get('method', 'unspecified'))
    for node, data in graph.nodes(data=True):
        safe.add_node(node, **{key: str(value) if isinstance(value, (set, list, dict)) else value for key, value in data.items() if key not in {'pages', 'domains'}})
    for source, target, data in graph.edges(data=True):
        safe.add_edge(source, target, **{key: str(value) if isinstance(value, (set, list, dict)) else value for key, value in data.items() if key not in {'pages', 'domains'}})
    nx.write_graphml(safe, assert_private_output_path(path))

def choose_cluster_count(matrix, maximum=10):
    from sklearn.cluster import KMeans
    from sklearn.metrics import silhouette_score
    if matrix.shape[0] < 3:
        return 1, pd.DataFrame(columns=['k', 'tfidf_silhouette'])
    trials = []
    for k in range(2, min(maximum, matrix.shape[0] - 1) + 1):
        trials.append((k, float(silhouette_score(matrix, KMeans(n_clusters=k, n_init=20, random_state=CLUSTER_RANDOM_STATE).fit_predict(matrix), metric='cosine'))))
    return max(trials, key=lambda item: item[1])[0], pd.DataFrame(trials, columns=['k', 'tfidf_silhouette'])

def model_token_chunks(text, model):
    token_ids = model.tokenizer(text, add_special_tokens=False, truncation=False, verbose=False)['input_ids']
    maximum = min(EMBEDDING_CHUNK_TOKENS, max(model.max_seq_length - 2, 8))
    overlap = min(EMBEDDING_CHUNK_OVERLAP_TOKENS, maximum - 1)
    if not token_ids:
        return ['']
    chunks, start = [], 0
    while start < len(token_ids):
        stop = min(start + maximum, len(token_ids))
        chunks.append(model.tokenizer.decode(token_ids[start:stop], skip_special_tokens=True))
        if stop == len(token_ids):
            break
        start = stop - overlap
    return chunks

def sentence_embeddings(texts, model_name):
    import numpy as np, torch
    from sentence_transformers import SentenceTransformer
    model = SentenceTransformer(model_name, device=EMBEDDING_RUNTIME['device'])
    model.eval()
    chunks, owner_index = [], []
    for index, text in enumerate(texts):
        page_chunks = model_token_chunks(text, model)
        chunks.extend(page_chunks)
        owner_index.extend([index] * len(page_chunks))
    effective_batch_size = min(EMBEDDING_RUNTIME['batch_size'], max(len(chunks), 1))
    with torch.inference_mode():
        chunk_embeddings = model.encode(chunks, batch_size=effective_batch_size, show_progress_bar=False, normalize_embeddings=True, convert_to_numpy=True)
    if EMBEDDING_RUNTIME['device'] == 'cuda':
        torch.cuda.synchronize()
        EMBEDDING_RUNTIME['peak_allocated_gib'] = round(max(EMBEDDING_RUNTIME.get('peak_allocated_gib', 0.0), torch.cuda.max_memory_allocated() / 1024**3), 3)
    document_embeddings, chunk_counts = np.zeros((len(texts), chunk_embeddings.shape[1]), dtype=np.float32), np.zeros(len(texts), dtype=int)
    for vector, index in zip(chunk_embeddings, owner_index):
        document_embeddings[index] += vector
        chunk_counts[index] += 1
    document_embeddings /= np.maximum(chunk_counts[:, None], 1)
    document_embeddings /= np.maximum(np.linalg.norm(document_embeddings, axis=1, keepdims=True), 1e-12)
    diagnostics = pd.DataFrame({'document_index': range(len(texts)), 'model': model_name, 'chunk_count': chunk_counts, 'device': EMBEDDING_RUNTIME['device'], 'batch_size': effective_batch_size, 'model_max_seq_length': model.max_seq_length})
    if EMBEDDING_RUNTIME['device'] == 'cuda':
        model.to('cpu')
        del model
        torch.cuda.empty_cache()
    return document_embeddings, diagnostics

def run_established_baseline(acquisition_result):
    import numpy as np
    from sklearn.cluster import KMeans
    from sklearn.metrics.pairwise import cosine_similarity
    if acquisition_result is None:
        raise RuntimeError('Stage 7 requires an acquisition snapshot.')
    ANALYSIS_DIR.mkdir(parents=True, exist_ok=True)
    documents = prepare_analysis_documents(acquisition_result['pages'])
    if len(documents) < 3:
        raise RuntimeError('At least three analysable pages are required.')
    vectorizer, tfidf_matrix = fit_tfidf(documents)
    raw_candidates = raw_phrase_candidate_table(documents, vectorizer, tfidf_matrix)
    review_ledger = build_concept_review_ledger(raw_candidates)
    raw_graph = build_raw_phrase_clique_graph(raw_candidates)
    raw_nodes, raw_edges, _ = graph_tables(raw_graph, 'raw_page_wide_tfidf_clique_diagnostic')
    review_ledger = enrich_concept_review_ledger(review_ledger, raw_nodes)
    review_batch_summary = build_review_batch_summary(review_ledger)
    community_review_summary = build_community_review_summary(review_ledger)
    provenance = concept_provenance_table(raw_candidates, review_ledger)
    bipartite_graph = build_page_concept_bipartite_graph(provenance)
    relationship_candidates = relationship_candidate_table(documents, provenance)
    substantive_graph = build_substantive_relationship_graph(provenance, relationship_candidates)
    substantive_nodes, substantive_edges, _ = graph_tables(substantive_graph, 'reviewed_local_cooccurrence_with_independent_recurrence')
    texts = documents['raw_text'].tolist()
    bert, bert_chunking = sentence_embeddings(texts, BERT_MODEL_NAME)
    roberta, roberta_chunking = sentence_embeddings(texts, ROBERTA_MODEL_NAME)
    embedding_chunking = pd.concat([bert_chunking, roberta_chunking], ignore_index=True)
    k, k_trials = choose_cluster_count(tfidf_matrix)
    representations = {'tfidf': tfidf_matrix, 'bert': bert, 'roberta': roberta}
    clusters, similarities = documents[['page_id', 'url', 'domain']].copy(), {}
    for name, matrix in representations.items():
        clusters[f'{name}_cluster'] = KMeans(n_clusters=k, n_init=20, random_state=CLUSTER_RANDOM_STATE).fit_predict(matrix) if k > 1 else np.zeros(len(documents), dtype=int)
        similarities[name] = cosine_similarity(matrix)
        np.save(assert_private_output_path(ANALYSIS_DIR / f'{name}_embeddings.npy'), matrix.toarray() if hasattr(matrix, 'toarray') else matrix)
        np.save(assert_private_output_path(ANALYSIS_DIR / f'{name}_similarity.npy'), similarities[name])
    documents.drop(columns=['raw_text', 'token_text']).to_csv(assert_private_output_path(ANALYSIS_DIR / 'documents.csv'), index=False)
    raw_candidates.to_csv(assert_private_output_path(ANALYSIS_DIR / 'raw_phrase_candidates.csv'), index=False)
    review_ledger.to_csv(assert_private_output_path(CONCEPT_REVIEW_LEDGER_FILE), index=False)
    review_batch_summary.to_csv(assert_private_output_path(ANALYSIS_DIR / 'review_batch_summary.csv'), index=False)
    community_review_summary.to_csv(assert_private_output_path(ANALYSIS_DIR / 'community_review_summary.csv'), index=False)
    provenance.to_csv(assert_private_output_path(CONCEPT_PROVENANCE_FILE), index=False)
    relationship_candidates.to_csv(assert_private_output_path(RELATIONSHIP_CANDIDATES_FILE), index=False)
    raw_nodes.to_csv(assert_private_output_path(ANALYSIS_DIR / 'raw_phrase_nodes.csv'), index=False)
    raw_edges.to_csv(assert_private_output_path(ANALYSIS_DIR / 'raw_phrase_edges.csv'), index=False)
    substantive_nodes.to_csv(assert_private_output_path(ANALYSIS_DIR / 'substantive_concept_nodes.csv'), index=False)
    substantive_edges.to_csv(assert_private_output_path(ANALYSIS_DIR / 'substantive_concept_edges.csv'), index=False)
    clusters.to_csv(assert_private_output_path(ANALYSIS_DIR / 'page_clusters.csv'), index=False)
    embedding_chunking.to_csv(assert_private_output_path(ANALYSIS_DIR / 'embedding_chunking.csv'), index=False)
    write_json(ANALYSIS_DIR / 'embedding_runtime.json', EMBEDDING_RUNTIME)
    k_trials.to_csv(assert_private_output_path(ANALYSIS_DIR / 'cluster_count_trials.csv'), index=False)
    write_graphml_safely(raw_graph, ANALYSIS_DIR / 'raw_phrase_clique_diagnostic.graphml')
    write_graphml_safely(bipartite_graph, ANALYSIS_DIR / 'reviewed_page_concept_bipartite.graphml')
    write_graphml_safely(substantive_graph, ANALYSIS_DIR / 'substantive_relationship_graph.graphml')
    review_counts = review_ledger['decision'].value_counts().to_dict() if not review_ledger.empty else {}
    batch_counts = review_ledger['review_batch'].value_counts().to_dict() if 'review_batch' in review_ledger.columns and not review_ledger.empty else {}
    review_gate = {'mode': CONCEPT_REVIEW_MODE, 'allow_unreviewed_concepts': ALLOW_UNREVIEWED_CONCEPTS, 'decision_counts': review_counts, 'batch_counts': batch_counts, 'pending_candidates': int(review_counts.get('pending', 0)), 'accepted_candidates': int(review_counts.get('accept', 0) + review_counts.get('rename', 0)), 'high_priority_candidates': int(batch_counts.get('batch_1_high_priority_target_aligned', 0)), 'substantive_graph_ready': bool(substantive_graph.number_of_edges())}
    summary = {'status': 'complete', 'snapshot_id': SNAPSHOT_ID, 'documents': len(documents), 'tfidf_features': int(tfidf_matrix.shape[1]), 'raw_phrase_nodes': int(raw_graph.number_of_nodes()), 'raw_phrase_edges': int(raw_graph.number_of_edges()), 'substantive_concept_nodes': int(substantive_graph.number_of_nodes()), 'substantive_concept_edges': int(substantive_graph.number_of_edges()), 'selected_page_clusters': int(k), 'models': EMBEDDING_MODEL_METADATA, 'embedding_runtime': EMBEDDING_RUNTIME, 'concept_review_gate': review_gate, 'method_status': 'raw_phrase_graph_is_diagnostic_only; substantive_graph_requires_reviewed_concepts_and_independent_recurrence'}
    write_json(ANALYSIS_DIR / 'stage7_baseline_summary.json', summary)
    write_checkpoint('stage7_established_methods', 'complete', **{key: value for key, value in summary.items() if key != 'status'})
    return {'documents': documents, 'tfidf_vectorizer': vectorizer, 'tfidf_matrix': tfidf_matrix, 'raw_candidates': raw_candidates, 'review_ledger': review_ledger, 'provenance': provenance, 'relationship_candidates': relationship_candidates, 'review_batch_summary': review_batch_summary, 'community_review_summary': community_review_summary, 'bipartite_graph': bipartite_graph, 'raw_graph': raw_graph, 'raw_nodes': raw_nodes, 'raw_edges': raw_edges, 'substantive_graph': substantive_graph, 'substantive_nodes': substantive_nodes, 'substantive_edges': substantive_edges, 'clusters': clusters, 'embedding_chunking': embedding_chunking, 'representations': representations, 'similarities': similarities, 'cluster_count': k, 'cluster_trials': k_trials, 'summary': summary}

stage7 = run_established_baseline(acquisition) if RUN_BASELINE_ANALYSIS else None
if stage7:
    display(pd.DataFrame([stage7['summary']]))
    display(stage7['review_ledger'].head(25))
else:
    print('Stage 7 prepared. Its private execution gate is off.')
````

````python
def extraction_sensitivity(pages):
    import trafilatura
    rows = []
    for page in pages:
        relative = page.get('private_source_path')
        if not relative:
            continue
        source_path = assert_private_output_path(OUTPUT_DIR / relative)
        if not source_path.exists():
            continue
        html = source_path.read_text(encoding='utf-8', errors='replace')
        alternate = trafilatura.extract(html, include_comments=False, include_tables=True) or ''
        baseline = page.get('clean_body', '')
        baseline_tokens, alternate_tokens = set(re.findall(r'[a-z]{3,}', baseline.lower())), set(re.findall(r'[a-z]{3,}', alternate.lower()))
        union = baseline_tokens | alternate_tokens
        rows.append({'page_id': page['page_id'], 'url': page.get('resolved_identity_url') or page.get('final_fetch_url'), 'beautifulsoup_characters': len(baseline), 'trafilatura_characters': len(alternate), 'length_ratio_trafilatura_to_beautifulsoup': len(alternate) / max(len(baseline), 1), 'token_jaccard': len(baseline_tokens & alternate_tokens) / max(len(union), 1), 'trafilatura_empty': not bool(alternate.strip())})
    return pd.DataFrame(rows)

def upper_triangle_values(matrix):
    import numpy as np
    return matrix[np.triu_indices_from(matrix, k=1)]

def neighbour_overlap(left, right, neighbours=5):
    import numpy as np
    k = min(neighbours, left.shape[0] - 1)
    if k < 1:
        return 0.0
    values = []
    for index in range(left.shape[0]):
        left_order = [item for item in np.argsort(-left[index]) if item != index][:k]
        right_order = [item for item in np.argsort(-right[index]) if item != index][:k]
        values.append(len(set(left_order) & set(right_order)) / k)
    return float(np.mean(values))

def representation_diagnostics(stage7_result):
    from scipy.stats import spearmanr
    from sklearn.metrics import adjusted_rand_score, silhouette_score
    names, pair_rows = list(stage7_result['similarities']), []
    for left_index, left in enumerate(names):
        for right in names[left_index + 1:]:
            pair_rows.append({'left': left, 'right': right, 'similarity_spearman': float(spearmanr(upper_triangle_values(stage7_result['similarities'][left]), upper_triangle_values(stage7_result['similarities'][right])).statistic), 'top5_neighbour_jaccard': neighbour_overlap(stage7_result['similarities'][left], stage7_result['similarities'][right]), 'cluster_ari': float(adjusted_rand_score(stage7_result['clusters'][f'{left}_cluster'], stage7_result['clusters'][f'{right}_cluster']))})
    silhouettes = {}
    if stage7_result['cluster_count'] > 1:
        for name, matrix in stage7_result['representations'].items():
            silhouettes[name] = float(silhouette_score(matrix, stage7_result['clusters'][f'{name}_cluster'], metric='cosine'))
    return pd.DataFrame(pair_rows), silhouettes

def kmeans_seed_stability(stage7_result, seeds=(11, 23, 42, 71, 101)):
    import itertools
    from sklearn.cluster import KMeans
    from sklearn.metrics import adjusted_rand_score
    if stage7_result['cluster_count'] < 2:
        return pd.DataFrame(columns=['representation', 'mean_pairwise_ari', 'minimum_pairwise_ari'])
    rows = []
    for name, matrix in stage7_result['representations'].items():
        assignments = {seed: KMeans(n_clusters=stage7_result['cluster_count'], n_init=1, random_state=seed).fit_predict(matrix) for seed in seeds}
        scores = [adjusted_rand_score(assignments[left], assignments[right]) for left, right in itertools.combinations(seeds, 2)]
        rows.append({'representation': name, 'mean_pairwise_ari': float(sum(scores) / len(scores)), 'minimum_pairwise_ari': float(min(scores))})
    return pd.DataFrame(rows)

def louvain_stability(graph, seeds=(11, 42, 71), resolutions=(0.75, 1.0, 1.25)):
    import itertools
    import community as community_louvain
    from sklearn.metrics import adjusted_rand_score
    columns = ['run', 'communities', 'modularity']
    if graph.number_of_nodes() < 2 or graph.number_of_edges() == 0:
        return pd.DataFrame(columns=columns), {'applicable': False, 'reason': 'Graph has fewer than two connected concepts.', 'mean_pairwise_ari': None, 'minimum_pairwise_ari': None}
    nodes, runs, metadata = sorted(graph.nodes()), {}, []
    for resolution in resolutions:
        for seed in seeds:
            partition = community_louvain.best_partition(graph, weight='weight', resolution=resolution, random_state=seed)
            key = f'resolution={resolution}|seed={seed}'
            runs[key] = [partition[node] for node in nodes]
            metadata.append({'run': key, 'communities': len(set(partition.values())), 'modularity': float(community_louvain.modularity(partition, graph, weight='weight'))})
    scores = [adjusted_rand_score(runs[left], runs[right]) for left, right in itertools.combinations(runs, 2)]
    return pd.DataFrame(metadata, columns=columns), {'applicable': True, 'mean_pairwise_ari': float(sum(scores) / len(scores)), 'minimum_pairwise_ari': float(min(scores))}

def graph_reliability(nodes, edges, graph):
    import networkx as nx
    concepts = set(nodes['concept']) if not nodes.empty else set()
    surface_pairs = sorted({(concept, concept[:-1]) for concept in concepts if concept.endswith('s') and len(concept) > 4 and concept[:-1] in concepts})
    return {'node_count': int(len(nodes)), 'edge_count': int(len(edges)), 'singleton_node_ratio': float((nodes['page_support'] == 1).mean()) if len(nodes) else None, 'single_page_edge_ratio': float((edges['page_support'] == 1).mean()) if len(edges) else None, 'connected_components': int(nx.number_connected_components(graph)) if graph.number_of_nodes() else 0, 'largest_component_share': max((len(component) for component in nx.connected_components(graph)), default=0) / max(graph.number_of_nodes(), 1), 'surface_plural_singular_pairs': [list(pair) for pair in surface_pairs[:100]], 'surface_plural_singular_pair_count': len(surface_pairs)}

def review_gate_summary(stage7_result):
    ledger = stage7_result['review_ledger']
    decisions = ledger['decision'].value_counts().to_dict() if not ledger.empty else {}
    return {'candidate_count': int(len(ledger)), 'decision_counts': decisions, 'pending_candidates': int(decisions.get('pending', 0)), 'accepted_or_renamed': int(decisions.get('accept', 0) + decisions.get('rename', 0)), 'substantive_graph_ready': bool(stage7_result['substantive_graph'].number_of_edges()), 'next_required_action': 'Human review of concept_review_ledger.csv before substantive relationship claims or larger acquisition.'}




HUMAN_PROVISIONAL_COMMUNITY_LABELS = {
    -1: "disregard / noise",
    0: "automating page SEO using Selenium",
    1: "using machine learning and Python to automate SEO",
    2: "using Claude and RankBrain for semantic SEO; using Claude to understand the RankBrain / Semantic SEO relationship",
    3: "internal linking strategy",
    4: "competitor gap analysis using Ahrefs to create new blog content",
    5: "commercial SEO offerings",
    6: "keyword clustering",
    7: "how RoBERTa works",
    8: "how spaCy works",
    9: "how BERT works",
    10: "how NLP is carried out",
    11: "how NetworkX works",
    12: "using spaCy to analyse ecommerce landing pages",
    13: "how to use Python for NLP and Semantic SEO using BERTopic",
}

def domain_cap_sensitivity_analysis(stage7_result, acquisition_result):
    """Post-hoc sensitivity: preserve the uncapped SERP, then test reduced same-domain views."""
    hits = acquisition_result.get('query_hits', pd.DataFrame()).copy() if acquisition_result else pd.DataFrame()
    raw_candidates = stage7_result.get('raw_candidates', pd.DataFrame()).copy()
    if hits.empty:
        return pd.DataFrame(), pd.DataFrame()
    hits['domain'] = hits['candidate_url'].map(registered_domain)
    hits = hits.sort_values(['query', 'rank', 'candidate_url']).copy()
    view_rows, concept_rows = [], []
    for cap in DOMAIN_CAP_SENSITIVITY:
        label = 'uncapped' if cap is None else f'cap_{cap}'
        if cap is None:
            kept_hits = hits.copy()
        else:
            kept_hits = hits.groupby(['query', 'domain'], group_keys=False).head(cap).copy()
        kept_page_ids = set(kept_hits.get('page_id', pd.Series(dtype=str)).dropna().astype(str))
        candidates = raw_candidates[raw_candidates['page_id'].astype(str).isin(kept_page_ids)].copy() if not raw_candidates.empty and kept_page_ids else pd.DataFrame()
        concept_counts = candidates['raw_surface'].astype(str).str.lower().value_counts() if not candidates.empty and 'raw_surface' in candidates.columns else pd.Series(dtype=int)
        view_rows.append({
            'view': label,
            'domain_cap': cap if cap is not None else 'uncapped',
            'query_hits': int(len(kept_hits)),
            'unique_domains': int(kept_hits['domain'].nunique()) if 'domain' in kept_hits.columns else 0,
            'unique_pages': int(len(kept_page_ids)),
            'raw_candidate_mentions': int(len(candidates)),
            'unique_raw_candidates': int(concept_counts.size),
            'top_candidates': ' | '.join(concept_counts.head(12).index.tolist()),
        })
        for concept, count in concept_counts.head(100).items():
            concept_rows.append({'view': label, 'concept': concept, 'mentions': int(count)})
    view_frame = pd.DataFrame(view_rows)
    concept_frame = pd.DataFrame(concept_rows)
    view_frame.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_domain_cap_sensitivity.csv'), index=False)
    concept_frame.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_domain_cap_concept_persistence.csv'), index=False)
    return view_frame, concept_frame


def _top_joined_terms(terms, limit=10):
    cleaned = []
    seen = set()
    for term in terms:
        text = str(term).strip()
        if not text or text.lower() == 'nan':
            continue
        key = text.lower()
        if key in seen:
            continue
        seen.add(key)
        cleaned.append(text)
        if len(cleaned) >= limit:
            break
    return ' | '.join(cleaned)


def _ctfidf_terms(assignments, term_column='raw_surface', group_column='group', limit=10):
    from collections import Counter, defaultdict
    frame = assignments[[group_column, term_column]].dropna().copy()
    if frame.empty:
        return {}
    class_counts = {}
    document_frequency = Counter()
    for group, rows in frame.groupby(group_column):
        counts = Counter(rows[term_column].astype(str).str.lower().str.strip())
        counts = Counter({term: count for term, count in counts.items() if term and term != 'nan'})
        class_counts[group] = counts
        for term in counts:
            document_frequency[term] += 1
    class_count = max(1, len(class_counts))
    labels = {}
    for group, counts in class_counts.items():
        total = max(1, sum(counts.values()))
        scored = []
        for term, count in counts.items():
            tf = count / total
            idf = np.log(1 + class_count / max(1, document_frequency[term]))
            scored.append((term, tf * idf, count))
        labels[group] = ' | '.join(term for term, _, _ in sorted(scored, key=lambda item: (item[1], item[2]), reverse=True)[:limit])
    return labels


def _tfidf_terms_by_group(assignments, group_column='group', limit=10):
    frame = assignments.dropna(subset=[group_column, 'raw_surface']).copy()
    if frame.empty:
        return {}
    frame['tfidf'] = pd.to_numeric(frame.get('tfidf', 0), errors='coerce').fillna(0)
    labels = {}
    for group, rows in frame.groupby(group_column):
        terms = rows.groupby('raw_surface')['tfidf'].sum().sort_values(ascending=False).index.tolist()
        labels[group] = _top_joined_terms(terms, limit=limit)
    return labels


def _community_label_rows(permutation, detection_method, label_method, labels, units, status, limitation, human_notes=None):
    rows = []
    for group in sorted(labels.keys(), key=lambda value: (str(type(value)), value)):
        group_units = units.get(group, {}) if isinstance(units, dict) else {}
        rows.append({
            'permutation': permutation,
            'community_detection': detection_method,
            'keyword_detection': label_method,
            'group_id': group,
            'label_terms': labels.get(group, ''),
            'human_provisional_label': human_notes.get(group, '') if human_notes else '',
            'page_count': group_units.get('page_count'),
            'domain_count': group_units.get('domain_count'),
            'status': status,
            'limitation': limitation,
        })
    return rows


def method_label_permutation_analysis(stage7_result):
    """Compare required community/topic detection methods against keyword-within-community labelling methods."""
    from sklearn.cluster import KMeans
    rows = []
    raw_nodes = stage7_result.get('raw_nodes', pd.DataFrame()).copy()
    raw_candidates = stage7_result.get('raw_candidates', pd.DataFrame()).copy()
    clusters = stage7_result.get('clusters', pd.DataFrame()).copy()
    documents = stage7_result.get('documents', pd.DataFrame()).copy()
    representations = stage7_result.get('representations', {})
    cluster_count = max(2, int(stage7_result.get('cluster_count', 8) or 8))

    if not raw_nodes.empty and {'concept', 'community'}.issubset(raw_nodes.columns):
        raw_nodes['community'] = pd.to_numeric(raw_nodes['community'], errors='coerce').fillna(-1).astype(int)
        louvain_units = {}
        for community, group in raw_nodes.groupby('community'):
            louvain_units[int(community)] = {
                'page_count': int(pd.to_numeric(group.get('page_support', 0), errors='coerce').fillna(0).sum()),
                'domain_count': int(pd.to_numeric(group.get('domain_support', 0), errors='coerce').fillna(0).sum()),
            }
        ranked = raw_nodes.sort_values(['community', 'page_support', 'weighted_degree'], ascending=[True, False, False])
        louvain_tfidf_labels = {int(community): _top_joined_terms(group['concept'].tolist(), 10) for community, group in ranked.groupby('community')}
        rows.extend(_community_label_rows(
            'NetworkX-Louvain Ã— TF-IDF',
            'NetworkX co-occurrence graph + Louvain community detection',
            'TF-IDF top phrases ranked by page support and graph degree',
            louvain_tfidf_labels,
            louvain_units,
            'run',
            'Uses the raw co-occurrence graph; communities are diagnostic until reviewed.',
            HUMAN_PROVISIONAL_COMMUNITY_LABELS,
        ))

        if not raw_candidates.empty and 'raw_surface' in raw_candidates.columns:
            community_lookup = raw_nodes.set_index('concept')['community'].to_dict()
            assigned = raw_candidates.copy()
            assigned['group'] = assigned['raw_surface'].map(community_lookup)
            assigned = assigned.dropna(subset=['group'])
            assigned['group'] = assigned['group'].astype(int)
            louvain_ctfidf_labels = _ctfidf_terms(assigned, group_column='group', term_column='raw_surface', limit=10)
            rows.extend(_community_label_rows(
                'NetworkX-Louvain Ã— c-TF-IDF',
                'NetworkX co-occurrence graph + Louvain community detection',
                'c-TF-IDF terms within Louvain communities',
                louvain_ctfidf_labels,
                louvain_units,
                'run',
                'c-TF-IDF improves community-specific labelling but still depends on raw extracted candidate quality.',
                HUMAN_PROVISIONAL_COMMUNITY_LABELS,
            ))

    if not raw_candidates.empty and not clusters.empty:
        for representation_name, cluster_column, permutation_name in [
            ('bert', 'bert_cluster', 'KMeans-BERT Ã— TF-IDF'),
            ('roberta', 'roberta_cluster', 'KMeans-RoBERTa Ã— TF-IDF'),
        ]:
            if cluster_column in clusters.columns:
                assigned = raw_candidates.merge(clusters[['page_id', cluster_column, 'domain']], on='page_id', how='inner', suffixes=('', '_cluster'))
                assigned = assigned.rename(columns={cluster_column: 'group'})
                labels = _tfidf_terms_by_group(assigned, group_column='group', limit=10)
                units = {}
                for group, group_rows in assigned.groupby('group'):
                    units[int(group)] = {
                        'page_count': int(group_rows['page_id'].nunique()),
                        'domain_count': int(group_rows['domain'].nunique()) if 'domain' in group_rows.columns else None,
                    }
                rows.extend(_community_label_rows(
                    permutation_name,
                    f'KMeans on {representation_name.upper()} page embeddings',
                    'TF-IDF candidate phrases aggregated within embedding clusters',
                    {int(key): value for key, value in labels.items()},
                    units,
                    'run',
                    'Clusters pages by embedding proximity, then explains those clusters with extracted surface phrases.',
                ))

    bertopic_status = 'not run: BERTopic/HDBSCAN dependencies unavailable in current runtime'
    try:
        import hdbscan  # noqa: F401
        if 'roberta' in representations and not raw_candidates.empty:
            clusterer = hdbscan.HDBSCAN(min_cluster_size=max(5, len(documents) // 25), min_samples=2)
            labels = clusterer.fit_predict(representations['roberta'])
            bertopic_frame = documents[['page_id', 'domain']].copy()
            bertopic_frame['group'] = labels
            assigned = raw_candidates.merge(bertopic_frame, on='page_id', how='inner', suffixes=('', '_document'))
            if 'domain' not in assigned.columns and 'domain_document' in assigned.columns:
                assigned['domain'] = assigned['domain_document']
            assigned = assigned[assigned['group'] >= 0]
            topic_labels = _ctfidf_terms(assigned, group_column='group', term_column='raw_surface', limit=10)
            units = {int(group): {'page_count': int(rows_['page_id'].nunique()), 'domain_count': int(rows_['domain'].nunique())} for group, rows_ in assigned.groupby('group')}
            rows.extend(_community_label_rows(
                'BERTopic-HDBSCAN Ã— c-TF-IDF',
                'HDBSCAN on RoBERTa page embeddings',
                'c-TF-IDF terms within density-based semantic clusters',
                {int(key): value for key, value in topic_labels.items()},
                units,
                'run',
                'Approximates the BERTopic structure when HDBSCAN is installed; noise points are excluded from topic labels.',
            ))
            bertopic_status = 'run'
    except Exception as exc:
        bertopic_status = f'not run: {type(exc).__name__}: {exc}'
    if not any(row['permutation'] == 'BERTopic-HDBSCAN Ã— c-TF-IDF' for row in rows):
        rows.append({
            'permutation': 'BERTopic-HDBSCAN Ã— c-TF-IDF',
            'community_detection': 'BERTopic / HDBSCAN topic detection',
            'keyword_detection': 'c-TF-IDF topic words',
            'group_id': None,
            'label_terms': '',
            'human_provisional_label': '',
            'page_count': None,
            'domain_count': None,
            'status': bertopic_status,
            'limitation': 'Required permutation. If not run, inspect the status for the dependency or runtime error that prevented topic extraction.',
        })

    for community, label in HUMAN_PROVISIONAL_COMMUNITY_LABELS.items():
        critique = 'Disregarded as noise unless later recurrence or evidence rescues part of it.' if community == -1 else 'Useful provisional human interpretation; agent must test whether computed labels and source evidence support it.'
        rows.append({
            'permutation': 'Human labels Ã— agent critique',
            'community_detection': 'Human-labelled Louvain communities',
            'keyword_detection': 'Agent critique constrained by evidence',
            'group_id': community,
            'label_terms': '',
            'human_provisional_label': label,
            'page_count': None,
            'domain_count': None,
            'status': 'human supervision recorded',
            'limitation': critique,
        })

    result = pd.DataFrame(rows)
    column_order = ['permutation', 'community_detection', 'keyword_detection', 'group_id', 'label_terms', 'human_provisional_label', 'page_count', 'domain_count', 'status', 'limitation']
    return result[column_order]


def topic_structure_comparison(stage7_result):
    """Retain only a compact count-level note; interpretation now lives in method_label_permutation_analysis."""
    rows = []
    raw_nodes = stage7_result.get('raw_nodes', pd.DataFrame())
    clusters = stage7_result.get('clusters', pd.DataFrame())
    if not raw_nodes.empty and 'community' in raw_nodes.columns:
        rows.append({'method': 'NetworkX-Louvain', 'unit_clustered': 'candidate phrases', 'groups_detected': int(raw_nodes['community'].nunique()), 'status': 'count only', 'limitation': 'Counts alone are not useful for review; inspect method-label permutations.'})
    for column, method in [('bert_cluster', 'KMeans-BERT'), ('roberta_cluster', 'KMeans-RoBERTa'), ('tfidf_cluster', 'KMeans-TF-IDF')]:
        if column in clusters.columns:
            rows.append({'method': method, 'unit_clustered': 'pages', 'groups_detected': int(clusters[column].nunique()), 'status': 'count only', 'limitation': 'Counts alone are not useful for review; inspect method-label permutations.'})
    rows.append({'method': 'BERTopic-HDBSCAN', 'unit_clustered': 'pages or embeddings', 'groups_detected': None, 'status': 'optional dependency-gated', 'limitation': 'The permutation table records whether this method ran.'})
    return pd.DataFrame(rows)

def codex_permutation_assessments(method_label_permutations):
    """Concise agent interpretation after each required permutation."""
    frame = method_label_permutations.copy() if isinstance(method_label_permutations, pd.DataFrame) else pd.DataFrame()
    rows = []
    order = [
        'NetworkX-Louvain Ã— TF-IDF',
        'NetworkX-Louvain Ã— c-TF-IDF',
        'KMeans-BERT Ã— TF-IDF',
        'KMeans-RoBERTa Ã— TF-IDF',
        'BERTopic-HDBSCAN Ã— c-TF-IDF',
        'Human labels Ã— agent critique',
    ]
    for permutation in order:
        subset = frame[frame['permutation'].eq(permutation)].copy() if not frame.empty and 'permutation' in frame.columns else pd.DataFrame()
        if subset.empty:
            rows.append({
                'permutation': permutation,
                'codex_assessment': 'No output was produced for this permutation in the current run.',
                'review_prompt': 'Check whether the method should have run or whether the absence is dependency-gated.',
            })
            continue
        run_status = ' | '.join(sorted(set(subset['status'].dropna().astype(str).tolist()))) if 'status' in subset.columns else 'unknown'
        labels = []
        for _, row in subset.head(18).iterrows():
            human_value = row.get('human_provisional_label', '')
            human = '' if pd.isna(human_value) else str(human_value).strip()
            terms_value = row.get('label_terms', '')
            terms = '' if pd.isna(terms_value) else str(terms_value).strip()
            group_id = row.get('group_id')
            if human:
                labels.append(f"{group_id}: {human}")
            elif terms:
                labels.append(f"{group_id}: {terms}")
        label_summary = '; '.join(labels[:10])
        if permutation == 'NetworkX-Louvain Ã— TF-IDF':
            assessment = 'This is the clearest graph baseline. It exposes recognisable SERP regions: Python automation, Semantic SEO/search systems, internal linking, competitor gaps, keyword clustering, model-specific method pages, NLP, spaCy, BERT, RoBERTa and NetworkX. Its weakness is that raw TF-IDF phrase support can still preserve artefacts and overlapping fragments.'
            prompt = 'Use this as the first visual map. Confirm whether the visible communities match the search-space regions you see.'
        elif permutation == 'NetworkX-Louvain Ã— c-TF-IDF':
            assessment = 'This keeps the same Louvain community structure but changes how each community is named. It should reveal which terms are distinctive to each community rather than simply frequent overall. Disagreement with the TF-IDF label is useful because it shows where the community meaning is fragile.'
            prompt = 'Compare with the TF-IDF Louvain map: are the labels sharper, or are important SEO meanings lost?'
        elif permutation == 'KMeans-BERT Ã— TF-IDF':
            assessment = 'This clusters pages by BERT embedding proximity and then explains the clusters using extracted TF-IDF phrases. It is useful for testing whether page-level semantic similarity reproduces the graph communities, but it may group pages by broad language similarity rather than practical SEO intent.'
            prompt = 'Look for whether BERT clusters preserve the practical article regions or smooth them into generic semantic groups.'
        elif permutation == 'KMeans-RoBERTa Ã— TF-IDF':
            assessment = 'This repeats the embedding-cluster test with RoBERTa. Its value is comparative: if RoBERTa and BERT produce similar labelled regions, confidence rises; if they diverge, the topic map is model-sensitive and should not be treated as settled.'
            prompt = 'Compare against BERT: which communities recur, and which are model artefacts?'
        elif permutation == 'BERTopic-HDBSCAN Ã— c-TF-IDF':
            if run_status == 'run':
                assessment = 'This modern topic-modelling permutation has now run. It uses HDBSCAN on RoBERTa page embeddings, then c-TF-IDF to label density-based semantic clusters. Its value is that it can expose compact page-level topic regions without relying on the raw phrase clique graph.'
                prompt = 'Compare these density-based topics with the NetworkX-Louvain and KMeans maps: which topic regions recur, which collapse together, and which are newly visible?'
            else:
                assessment = f'This is a required modern topic-modelling permutation, but the current runtime status is: {run_status}. It remains part of the notebook design because BERTopic/HDBSCAN directly matches the community-detection plus c-TF-IDF labelling pattern.'
                prompt = 'Inspect the runtime status and decide whether the failed dependency/runtime condition needs correction before the next run.'
        else:
            assessment = 'This records the human provisional interpretation as its own comparison layer. It is not a computational method, but it is central to the experiment: the agent must explain whether computed labels support, challenge, or fail to capture the human-readable search-space map.'
            prompt = 'Use this layer to accept, rename, split, merge, or reject computed communities.'
        if label_summary:
            assessment = assessment + ' Current visible labels include: ' + label_summary
        rows.append({'permutation': permutation, 'codex_assessment': assessment, 'review_prompt': prompt})
    return pd.DataFrame(rows)


def run_stage8(stage7_result, acquisition_result):
    if stage7_result is None:
        raise RuntimeError('Stage 8 requires Stage 7 results.')
    extraction = extraction_sensitivity(acquisition_result['pages'])
    pairs, silhouettes = representation_diagnostics(stage7_result)
    seed_stability = kmeans_seed_stability(stage7_result)
    raw_louvain_runs, raw_louvain_summary = louvain_stability(stage7_result['raw_graph'])
    substantive_louvain_runs, substantive_louvain_summary = louvain_stability(stage7_result['substantive_graph'])
    topic_comparison = topic_structure_comparison(stage7_result)
    method_label_permutations = method_label_permutation_analysis(stage7_result)
    codex_assessments = codex_permutation_assessments(method_label_permutations)
    domain_cap_sensitivity, domain_cap_concept_persistence = domain_cap_sensitivity_analysis(stage7_result, acquisition_result)
    raw_graph_summary = graph_reliability(stage7_result['raw_nodes'], stage7_result['raw_edges'], stage7_result['raw_graph'])
    substantive_graph_summary = graph_reliability(stage7_result['substantive_nodes'], stage7_result['substantive_edges'], stage7_result['substantive_graph'])
    extraction_summary = {'pages_compared': len(extraction), 'median_token_jaccard': float(extraction['token_jaccard'].median()) if len(extraction) else None, 'median_length_ratio': float(extraction['length_ratio_trafilatura_to_beautifulsoup'].median()) if len(extraction) else None, 'empty_alternate_extractions': int(extraction['trafilatura_empty'].sum()) if len(extraction) else 0}
    diagnostics = {'status': 'complete', 'snapshot_id': SNAPSHOT_ID, 'extraction': extraction_summary, 'representation_pairs': pairs.to_dict(orient='records'), 'silhouette_by_representation': silhouettes, 'kmeans_seed_stability': seed_stability.to_dict(orient='records'), 'raw_phrase_graph_reliability': raw_graph_summary, 'raw_phrase_louvain_stability': raw_louvain_summary, 'substantive_graph_reliability': substantive_graph_summary, 'substantive_louvain_stability': substantive_louvain_summary, 'concept_review_gate': review_gate_summary(stage7_result), 'bounded_conclusion': 'The raw page-wide phrase clique remains a failure diagnostic. No substantive concept graph is claimed until human-reviewed concepts produce relationships with independent page and domain recurrence.'}
    extraction.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_extraction_sensitivity.csv'), index=False)
    pairs.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_representation_pairs.csv'), index=False)
    seed_stability.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_kmeans_seed_stability.csv'), index=False)
    topic_comparison.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_topic_structure_comparison.csv'), index=False)
    method_label_permutations.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_method_label_permutations.csv'), index=False)
    codex_assessments.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_codex_permutation_assessments.csv'), index=False)
    domain_cap_sensitivity.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_domain_cap_sensitivity.csv'), index=False)
    domain_cap_concept_persistence.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_domain_cap_concept_persistence.csv'), index=False)
    raw_louvain_runs.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_raw_louvain_runs.csv'), index=False)
    substantive_louvain_runs.to_csv(assert_private_output_path(ANALYSIS_DIR / 'stage8_substantive_louvain_runs.csv'), index=False)
    write_json(ANALYSIS_DIR / 'stage8_diagnostics.json', diagnostics)
    write_checkpoint('stage8_diagnostics', 'complete', diagnostics='analysis/stage8_diagnostics.json', substantive_graph_ready=diagnostics['concept_review_gate']['substantive_graph_ready'])
    return {'diagnostics': diagnostics, 'extraction': extraction, 'representation_pairs': pairs, 'seed_stability': seed_stability, 'topic_comparison': topic_comparison, 'method_label_permutations': method_label_permutations, 'codex_permutation_assessments': codex_assessments, 'domain_cap_sensitivity': domain_cap_sensitivity, 'domain_cap_concept_persistence': domain_cap_concept_persistence, 'raw_louvain_runs': raw_louvain_runs, 'substantive_louvain_runs': substantive_louvain_runs}

stage8 = run_stage8(stage7, acquisition) if stage7 else None
if stage8:
    display(pd.DataFrame([{'median_extraction_jaccard': stage8['diagnostics']['extraction']['median_token_jaccard'], 'raw_single_page_edge_ratio': stage8['diagnostics']['raw_phrase_graph_reliability']['single_page_edge_ratio'], 'pending_concept_review': stage8['diagnostics']['concept_review_gate']['pending_candidates'], 'substantive_graph_ready': stage8['diagnostics']['concept_review_gate']['substantive_graph_ready']}]))
    display(stage8['representation_pairs'])
    display(stage7['review_ledger'].head(25))
else:
    print('Stage 8 prepared. It executes immediately after the private Stage 7 baseline.')
````

````python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd
import textwrap
from sklearn.decomposition import PCA

VISUALISATION_DPI = 160
VISUALISATION_DIR = assert_private_output_path(ANALYSIS_DIR / "figures")
VISUALISATION_DIR.mkdir(parents=True, exist_ok=True)

RAW_GRAPH_REVIEW_NODE_LIMIT = int(os.getenv("SEO_LAB_RAW_GRAPH_REVIEW_NODE_LIMIT", "80"))
RAW_GRAPH_MIN_PAGE_SUPPORT = int(os.getenv("SEO_LAB_RAW_GRAPH_MIN_PAGE_SUPPORT", "2"))


def _maybe_show(fig):
    backend = plt.get_backend().lower()
    if "agg" not in backend:
        plt.show()
    plt.close(fig)


def _empty_figure(title, message):
    fig, ax = plt.subplots(figsize=(9, 4.5))
    ax.axis("off")
    ax.set_title(title)
    ax.text(0.5, 0.5, message, ha="center", va="center", wrap=True, fontsize=11)
    return fig


def _save_figure(fig, filename, title, caption, limitation, artifacts):
    target = assert_private_output_path(VISUALISATION_DIR / filename)
    fig.savefig(target, dpi=VISUALISATION_DPI, bbox_inches="tight")
    artifacts.append({
        "title": title,
        "path": str(target.relative_to(OUTPUT_DIR)),
        "caption": caption,
        "limitation": limitation,
    })
    _maybe_show(fig)
    return target


def _draw_substantive_graph(stage7_result):
    graph = stage7_result["substantive_graph"]
    nodes = stage7_result["substantive_nodes"].copy()
    if graph.number_of_nodes() == 0:
        return None

    fig, ax = plt.subplots(figsize=(12, 9))
    pos = nx.spring_layout(graph, seed=CLUSTER_RANDOM_STATE, weight="weight")
    widths = [0.75 + data.get("domain_support", 1) for _, _, data in graph.edges(data=True)]
    nx.draw_networkx_edges(graph, pos, ax=ax, width=widths, alpha=0.35, edge_color="#64748b")
    support = nodes.set_index("concept").reindex(list(graph.nodes())).fillna(1)
    sizes = 300 + 180 * support["page_support"].astype(float).to_numpy()
    nx.draw_networkx_nodes(graph, pos, ax=ax, node_size=sizes, node_color="#2563eb", alpha=0.82, edgecolors="#0f172a", linewidths=0.8)
    nx.draw_networkx_labels(graph, pos, ax=ax, font_size=8, font_color="#111827")
    ax.set_title("Evidence-filtered concept graph: every retained node is labelled")
    ax.axis("off")
    return fig



def _raw_graph_communities(graph):
    if graph.number_of_nodes() == 0:
        return []
    try:
        communities = list(nx.community.louvain_communities(graph, seed=CLUSTER_RANDOM_STATE, weight="weight"))
    except Exception:
        communities = [set(component) for component in nx.connected_components(graph)]
    return sorted((set(community) for community in communities), key=lambda community: (-len(community), sorted(community)[0] if community else ""))


def _write_raw_community_labels(stage7_result, communities):
    raw_nodes = stage7_result["raw_nodes"].copy()
    ledger = stage7_result.get("review_ledger", pd.DataFrame()).copy()
    node_lookup = raw_nodes.set_index("concept") if not raw_nodes.empty and "concept" in raw_nodes.columns else pd.DataFrame()
    ledger_lookup = ledger.set_index("proposed_canonical") if not ledger.empty and "proposed_canonical" in ledger.columns else pd.DataFrame()
    rows = []
    for community_id, community in enumerate(communities, start=1):
        for concept in sorted(community):
            record = {"raw_community": community_id, "concept": concept}
            if not node_lookup.empty and concept in node_lookup.index:
                for column in ["page_support", "domain_support", "pagerank", "weighted_degree"]:
                    if column in node_lookup.columns:
                        value = node_lookup.loc[concept, column]
                        record[column] = float(value) if column in {"pagerank", "weighted_degree"} else int(value)
            if not ledger_lookup.empty and concept in ledger_lookup.index:
                for column in ["review_batch", "review_priority", "suggested_action", "suggested_reason", "target_term_match", "automatic_flags", "likely_artifact"]:
                    if column in ledger_lookup.columns:
                        value = ledger_lookup.loc[concept, column]
                        record[column] = float(value) if column == "review_priority" else value
            rows.append(record)
    table = pd.DataFrame(rows)
    if not table.empty:
        table.to_csv(assert_private_output_path(ANALYSIS_DIR / "raw_community_labels.csv"), index=False)
    return table


def _community_label_figure(raw_community_labels):
    if raw_community_labels.empty:
        return _empty_figure("Raw community labels", "No raw community labels were available.")
    communities = []
    for community_id, group in raw_community_labels.groupby("raw_community"):
        ordered = group.sort_values([column for column in ["review_priority", "page_support", "domain_support", "weighted_degree"] if column in group.columns], ascending=False)
        concepts = []
        for row in ordered.head(14).itertuples(index=False):
            concept = getattr(row, "concept")
            action = getattr(row, "suggested_action", "") if hasattr(row, "suggested_action") else ""
            priority = getattr(row, "review_priority", "") if hasattr(row, "review_priority") else ""
            label = f"{concept}"
            if priority != "":
                label += f" [{priority}]"
            if action:
                label += f" - {action}"
            concepts.append(label)
        communities.append((community_id, concepts, len(group)))
    rows = min(4, max(1, len(communities)))
    cols = int(np.ceil(len(communities) / rows))
    fig, axes = plt.subplots(rows, cols, figsize=(6.4 * cols, 3.8 * rows), squeeze=False)
    for ax in axes.ravel():
        ax.axis("off")
    for ax, (community_id, concepts, total) in zip(axes.ravel(), communities):
        wrapped = []
        for concept in concepts:
            wrapped.extend(textwrap.wrap(concept, width=42) or [concept])
        body = "\n".join("- " + line for line in wrapped[:30])
        if total > len(concepts):
            body += f"\n- ... {total - len(concepts)} more in audit table"
        ax.text(0, 1, f"Raw community {community_id} ({total} candidates)\n{body}", va="top", ha="left", fontsize=7.5, family="monospace")
    fig.suptitle("Raw diagnostic communities: review-prioritised labels, not validated topics")
    fig.tight_layout(rect=(0, 0, 1, 0.96))
    return fig




def _domain_concentration_figure(acquisition_result):
    hits = pd.DataFrame(acquisition_result.get("query_hits", [])) if acquisition_result else pd.DataFrame()
    if hits.empty or "candidate_url" not in hits.columns:
        return _empty_figure("Domain concentration", "No query-hit table was available.")
    hits = hits.copy()
    hits["domain"] = hits["candidate_url"].map(registered_domain)
    rows = []
    for query, group in hits.groupby("query"):
        counts = group["domain"].value_counts()
        rows.append({
            "query": query,
            "top_domain_share": float(counts.iloc[0] / len(group)) if len(counts) else 0.0,
            "duplicate_domain_share": float((len(group) - counts.size) / len(group)) if len(group) else 0.0,
            "top_domain": counts.index[0] if len(counts) else "",
        })
    frame = pd.DataFrame(rows).sort_values("top_domain_share", ascending=True)
    fig, ax = plt.subplots(figsize=(11, max(4, 0.48 * len(frame))))
    y = np.arange(len(frame))
    ax.barh(y - 0.18, frame["top_domain_share"], height=0.34, label="Top-domain share", color="#2563eb")
    ax.barh(y + 0.18, frame["duplicate_domain_share"], height=0.34, label="Duplicate-domain share", color="#7c3aed")
    ax.set_yticks(y)
    ax.set_yticklabels(frame["query"])
    ax.set_xlim(0, 1)
    ax.set_xlabel("Share of returned results")
    ax.set_title("Domain concentration by query in the uncapped SERP")
    for index, row in enumerate(frame.itertuples(index=False)):
        ax.text(min(row.top_domain_share + 0.015, 0.98), index - 0.18, row.top_domain, va="center", fontsize=7)
    ax.legend(loc="lower right")
    fig.tight_layout()
    return fig


def _domain_cap_sensitivity_figure(stage8_result):
    frame = stage8_result.get("domain_cap_sensitivity", pd.DataFrame()).copy()
    if frame.empty:
        return _empty_figure("Domain cap sensitivity", "No domain-cap sensitivity table was available.")
    fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))
    frame.plot(kind="bar", x="view", y=["query_hits", "unique_pages"], ax=axes[0])
    axes[0].set_title("SERP evidence retained by cap view")
    axes[0].set_ylabel("Count")
    axes[0].tick_params(axis="x", rotation=0)
    frame.plot(kind="bar", x="view", y=["unique_domains", "unique_raw_candidates"], ax=axes[1])
    axes[1].set_title("Diversity retained by cap view")
    axes[1].set_ylabel("Count")
    axes[1].tick_params(axis="x", rotation=0)
    fig.suptitle("Post-hoc domain cap sensitivity: uncapped first, capped views second")
    fig.tight_layout(rect=(0, 0, 1, 0.93))
    return fig

def _community_count_stability_figure(stage8_result):
    runs = stage8_result.get("raw_louvain_runs", pd.DataFrame()).copy()
    if runs.empty or "run" not in runs.columns:
        return _empty_figure("Community count stability", "No Louvain stability runs were available.")
    extracted = runs["run"].str.extract(r"resolution=(?P<resolution>[0-9.]+)\|seed=(?P<seed>[0-9]+)")
    runs["resolution"] = extracted["resolution"].astype(float)
    runs["seed"] = extracted["seed"].astype(int)
    summary = runs.groupby("resolution").agg(
        min_communities=("communities", "min"),
        max_communities=("communities", "max"),
        mean_communities=("communities", "mean"),
        mean_modularity=("modularity", "mean"),
    ).reset_index()
    fig, ax1 = plt.subplots(figsize=(9.5, 4.8))
    ax1.plot(summary["resolution"], summary["mean_communities"], marker="o", color="#2563eb")
    ax1.fill_between(summary["resolution"], summary["min_communities"], summary["max_communities"], color="#bfdbfe", alpha=0.6)
    for row in summary.itertuples(index=False):
        ax1.text(row.resolution, row.mean_communities + 0.05, f"{row.mean_communities:.0f}", ha="center", va="bottom")
    ax1.set_title("Raw Louvain community count changes with resolution")
    ax1.set_xlabel("Louvain resolution")
    ax1.set_ylabel("Communities detected")
    ax1.set_ylim(0, max(summary["max_communities"].max() + 1, 2))
    ax2 = ax1.twinx()
    ax2.plot(summary["resolution"], summary["mean_modularity"], marker="s", color="#be123c", alpha=0.75)
    ax2.set_ylabel("Mean modularity")
    fig.tight_layout()
    return fig


def _community_composition_figure(community_summary):
    if community_summary.empty:
        return _empty_figure("Community composition", "No community summary was available.")
    rows = len(community_summary)
    fig, axes = plt.subplots(rows, 1, figsize=(13, max(2.75 * rows, 5.5)), squeeze=False)
    for ax, row in zip(axes.ravel(), community_summary.sort_values("raw_community").itertuples(index=False)):
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        ax.axis("off")
        community = getattr(row, "raw_community")
        count = getattr(row, "candidate_count")
        review_first = str(getattr(row, "review_first", "") or "None identified")
        likely_rejects = str(getattr(row, "likely_rejects", "") or "None identified")

        ax.text(0.00, 0.92, f"Community {community} ({count} candidates)", fontsize=11, weight="bold", va="top")
        ax.hlines(0.80, 0.00, 1.00, color="#cbd5e1", linewidth=1.2)

        ax.text(0.00, 0.68, "REVIEW FIRST", fontsize=8.5, weight="bold", color="#166534", va="top")
        ax.text(
            0.17,
            0.68,
            "\n".join(textwrap.wrap(review_first, width=112)),
            fontsize=8.3,
            family="monospace",
            va="top",
            color="#111827",
        )

        ax.hlines(0.41, 0.00, 1.00, color="#e2e8f0", linewidth=0.9, linestyle="--")

        ax.text(0.00, 0.29, "LIKELY REJECTS / ARTEFACTS", fontsize=8.5, weight="bold", color="#991b1b", va="top")
        ax.text(
            0.26,
            0.29,
            "\n".join(textwrap.wrap(likely_rejects, width=100)),
            fontsize=8.3,
            family="monospace",
            va="top",
            color="#111827",
        )
    fig.suptitle("Community composition: review candidates separated from likely rejects")
    fig.tight_layout(rect=(0, 0, 1, 0.95))
    return fig


def _page_community_matrix(stage7_result):
    raw_candidates = stage7_result.get("raw_candidates", pd.DataFrame()).copy()
    ledger = stage7_result.get("review_ledger", pd.DataFrame()).copy()
    documents = stage7_result.get("documents", pd.DataFrame()).copy()
    if raw_candidates.empty or ledger.empty or "raw_surface" not in raw_candidates.columns or "proposed_canonical" not in ledger.columns:
        return pd.DataFrame()
    lookup = ledger[["proposed_canonical", "raw_community"]].dropna().copy()
    lookup["join_term"] = lookup["proposed_canonical"].astype(str).str.lower()
    candidates = raw_candidates.copy()
    candidates["join_term"] = candidates["raw_surface"].astype(str).str.lower()
    merged = candidates.merge(lookup[["join_term", "raw_community"]], on="join_term", how="inner")
    if merged.empty:
        return pd.DataFrame()
    matrix = merged.groupby(["page_id", "raw_community"]).size().unstack(fill_value=0)
    if not documents.empty and {"page_id", "domain"}.issubset(documents.columns):
        labels = documents.set_index("page_id")["domain"].astype(str).to_dict()
        matrix.index = [labels.get(index, index) for index in matrix.index]
    matrix = matrix.sort_index(axis=1)
    matrix.to_csv(assert_private_output_path(ANALYSIS_DIR / "page_community_matrix.csv"))
    return matrix


def _page_community_matrix_figure(stage7_result):
    matrix = _page_community_matrix(stage7_result)
    if matrix.empty:
        return _empty_figure("Page-community evidence matrix", "No page-community matrix could be built from raw candidates and review ledger.")
    fig, ax = plt.subplots(figsize=(9.5, max(4, 0.55 * len(matrix))))
    im = ax.imshow(matrix.to_numpy(), aspect="auto", cmap="Blues")
    ax.set_xticks(range(matrix.shape[1]))
    ax.set_xticklabels([f"Community {column}" for column in matrix.columns], rotation=35, ha="right")
    ax.set_yticks(range(matrix.shape[0]))
    ax.set_yticklabels(matrix.index)
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            value = int(matrix.iloc[i, j])
            if value:
                ax.text(j, i, str(value), ha="center", va="center", fontsize=8)
    ax.set_title("Page Ã— raw community evidence matrix")
    fig.colorbar(im, ax=ax, shrink=0.8, label="Candidate phrase count")
    fig.tight_layout()
    return fig


def _topic_structure_comparison_figure(stage8_result):
    comparison = stage8_result.get("topic_comparison", pd.DataFrame()).copy()
    if comparison.empty:
        return _empty_figure("Topic structure comparison", "No topic-structure comparison table was available.")
    plot = comparison[comparison["groups_detected"].notna()].copy()
    if plot.empty:
        return _empty_figure("Topic structure comparison", "No method produced a comparable group count in this run.")
    fig, ax = plt.subplots(figsize=(11, 4.8))
    plot["groups_detected"] = plot["groups_detected"].astype(float)
    plot.plot(kind="barh", x="method", y="groups_detected", ax=ax, legend=False, color="#2563eb")
    ax.set_xlabel("Groups detected")
    ax.set_ylabel("Method")
    ax.set_title("Different methods produce different topic/community structures")
    for index, value in enumerate(plot["groups_detected"]):
        ax.text(value, index, f" {int(value)}", va="center")
    fig.tight_layout()
    return fig

def _review_batch_figure(review_batch_summary):
    if review_batch_summary.empty:
        return _empty_figure("Review batches", "No review batch summary was available.")
    fig, ax = plt.subplots(figsize=(10, 4.8))
    plot_data = review_batch_summary.copy().sort_values("review_batch")
    plot_data.plot(kind="barh", x="review_batch", y="candidate_count", ax=ax, legend=False, color="#2563eb")
    ax.set_title("Human review batches")
    ax.set_xlabel("Candidate concepts")
    ax.set_ylabel("Review batch")
    for index, value in enumerate(plot_data["candidate_count"]):
        ax.text(value, index, f" {int(value)}", va="center")
    fig.tight_layout()
    return fig

def _truncate_label(value, limit=56):
    text = str(value or '').replace('\n', ' ').strip()
    return text if len(text) <= limit else text[: limit - 1].rstrip() + '?'


def _split_terms(value, limit=4):
    terms = []
    for term in str(value or '').split('|'):
        term = term.strip()
        if term and term.lower() != 'nan':
            terms.append(term)
        if len(terms) >= limit:
            break
    return terms


def _permutation_network_figure(permutation_frame, permutation_name):
    frame = permutation_frame[permutation_frame['permutation'].eq(permutation_name)].copy()
    if frame.empty:
        return _empty_figure(permutation_name, 'No rows were produced for this permutation.')
    if frame['status'].astype(str).str.startswith('not run').all():
        message = frame['status'].dropna().astype(str).iloc[0] if len(frame) else 'Dependency-gated permutation.'
        return _empty_figure(permutation_name, message)

    graph = nx.Graph()
    root = permutation_name
    graph.add_node(root, node_type='method', label=permutation_name)
    visible = frame.copy().head(16)
    for _, row in visible.iterrows():
        group_id = row.get('group_id')
        group_node = f"group::{permutation_name}::{group_id}"
        human_value = row.get('human_provisional_label', '')
        label_value = row.get('label_terms', '')
        human_label = '' if pd.isna(human_value) else str(human_value).strip()
        label_terms = '' if pd.isna(label_value) else str(label_value).strip()
        main_label = human_label or label_terms or str(row.get('status', ''))
        graph.add_node(group_node, node_type='group', label=f"{group_id}: {_truncate_label(main_label, 48)}")
        graph.add_edge(root, group_node, weight=2)
        for term in _split_terms(label_terms, limit=3):
            term_node = f"term::{permutation_name}::{group_id}::{term}"
            graph.add_node(term_node, node_type='term', label=_truncate_label(term, 32))
            graph.add_edge(group_node, term_node, weight=1)

    fig, ax = plt.subplots(figsize=(17, max(9, 0.55 * max(10, len(visible)))))
    pos = nx.spring_layout(graph, seed=CLUSTER_RANDOM_STATE, k=1.25, iterations=180, weight='weight')
    node_types = nx.get_node_attributes(graph, 'node_type')
    method_nodes = [node for node, node_type in node_types.items() if node_type == 'method']
    group_nodes = [node for node, node_type in node_types.items() if node_type == 'group']
    term_nodes = [node for node, node_type in node_types.items() if node_type == 'term']
    nx.draw_networkx_edges(graph, pos, ax=ax, width=0.8, alpha=0.22, edge_color='#64748b')
    if method_nodes:
        nx.draw_networkx_nodes(graph, pos, nodelist=method_nodes, node_size=1800, node_color='#111827', alpha=0.90, ax=ax)
    if group_nodes:
        sizes = []
        for node in group_nodes:
            group_id = node.split('::')[-1]
            match = visible[visible['group_id'].astype(str).eq(str(group_id))]
            page_count = pd.to_numeric(match.get('page_count', pd.Series([1])), errors='coerce').fillna(1).max() if not match.empty else 1
            sizes.append(430 + min(float(page_count), 250) * 4)
        nx.draw_networkx_nodes(graph, pos, nodelist=group_nodes, node_size=sizes, node_color='#3b82f6', alpha=0.74, edgecolors='#1e3a8a', linewidths=0.8, ax=ax)
    if term_nodes:
        nx.draw_networkx_nodes(graph, pos, nodelist=term_nodes, node_size=180, node_color='#93c5fd', alpha=0.62, edgecolors='#64748b', linewidths=0.35, ax=ax)
    labels = nx.get_node_attributes(graph, 'label')
    group_labels = {node: labels[node] for node in group_nodes}
    term_labels = {node: labels[node] for node in term_nodes}
    root_labels = {node: labels[node] for node in method_nodes}
    nx.draw_networkx_labels(graph, pos, labels=root_labels, ax=ax, font_size=9, font_color='white')
    nx.draw_networkx_labels(graph, pos, labels=group_labels, ax=ax, font_size=7, font_color='#111827', bbox={'boxstyle': 'round,pad=0.16', 'fc': 'white', 'ec': '#d1d5db', 'alpha': 0.86})
    nx.draw_networkx_labels(graph, pos, labels=term_labels, ax=ax, font_size=6, font_color='#334155', bbox={'boxstyle': 'round,pad=0.10', 'fc': 'white', 'ec': 'none', 'alpha': 0.72})
    ax.set_title(f'{permutation_name}: topic labels as a review network')
    ax.text(0.01, 0.01, 'Centre = method permutation; blue nodes = communities/topics; small nodes = top extracted terms. Full rows remain in the audit CSV.', transform=ax.transAxes, fontsize=9, color='#475569')
    ax.axis('off')
    return fig


def _save_permutation_networks(stage8_result, artifacts):
    frame = stage8_result.get('method_label_permutations', pd.DataFrame())
    if frame is None or frame.empty:
        return
    names = [
        'NetworkX-Louvain Ã— TF-IDF',
        'NetworkX-Louvain Ã— c-TF-IDF',
        'KMeans-BERT Ã— TF-IDF',
        'KMeans-RoBERTa Ã— TF-IDF',
        'BERTopic-HDBSCAN Ã— c-TF-IDF',
        'Human labels Ã— agent critique',
    ]
    for index, name in enumerate(names, start=1):
        fig = _permutation_network_figure(frame, name)
        slug = (name.lower()
            .replace('networkx-', 'networkx-')
            .replace(' Ã— ', '-x-')
            .replace(' ', '-')
            .replace('/', '-')
            .replace('?', '-')
            .replace('?', '-')
        )
        slug = ''.join(ch for ch in slug if ch.isalnum() or ch in '-_')
        _save_figure(
            fig,
            f'07{index}-permutation-{slug}.png',
            f'Permutation: {name}',
            'Network-style topic review for one community-detection Ã— keyword-detection permutation.',
            'This visual is for interpretation and comparison. It does not prove reliability; stability and evidence checks remain separate.',
            artifacts,
        )


def generate_stage8_visualisations(stage7_result, stage8_result, acquisition_result):
    if stage7_result is None or stage8_result is None:
        return []

    artifacts = []
    pages = pd.DataFrame(acquisition_result.get("pages", [])) if acquisition_result else pd.DataFrame()

    if not pages.empty and {"query", "status"}.issubset(pages.columns):
        status_counts = pages.groupby(["query", "status"]).size().reset_index(name="pages")
        pivot = status_counts.pivot(index="query", columns="status", values="pages").fillna(0)
        fig, ax = plt.subplots(figsize=(11, max(4, 0.55 * len(pivot))))
        pivot.plot(kind="barh", stacked=True, ax=ax, colormap="tab20")
        ax.set_title("Acquisition coverage by query and retrieval status")
        ax.set_xlabel("Returned pages")
        ax.set_ylabel("Query")
        ax.legend(title="Status", bbox_to_anchor=(1.02, 1), loc="upper left")
    else:
        fig = _empty_figure("Acquisition coverage", "No page-level acquisition table was available for this replay.")
    _save_figure(fig, "01-acquisition-coverage.png", "Acquisition coverage", "Shows which returned URLs produced inspectable text and which were deferred, blocked, or unavailable.", "The chart measures retrieval coverage, not ranking quality or search demand.", artifacts)

    fig = _domain_concentration_figure(acquisition_result)
    _save_figure(fig, "02-domain-concentration.png", "Domain concentration", "Shows whether the uncapped SERP is dominated by repeated domains before any semantic filtering is applied.", "Domain concentration is a property of the returned SERP, not proof of topical authority.", artifacts)

    ledger = stage7_result["review_ledger"].copy()
    if not ledger.empty and "decision" in ledger.columns:
        counts = ledger["decision"].fillna("pending").value_counts().reindex(["pending", "accept", "rename", "reject"], fill_value=0)
        fig, ax = plt.subplots(figsize=(8, 4.5))
        counts.plot(kind="bar", ax=ax, color=["#f59e0b", "#16a34a", "#0ea5e9", "#dc2626"])
        ax.set_title("Concept review gate")
        ax.set_ylabel("Candidate concepts")
        ax.set_xlabel("Review decision")
        ax.tick_params(axis="x", rotation=0)
        for index, value in enumerate(counts):
            ax.text(index, value, str(int(value)), ha="center", va="bottom")
    else:
        fig = _empty_figure("Concept review gate", "No concept review ledger was produced.")
    _save_figure(fig, "02-concept-review-gate.png", "Concept review gate", "Shows how many extracted phrase candidates are pending, accepted, renamed, or rejected before they may support substantive graph claims.", "Pending candidates are not evidence of meaningful concepts; they are work queued for human supervision.", artifacts)

    fig = _review_batch_figure(stage7_result.get("review_batch_summary", pd.DataFrame()))
    _save_figure(fig, "03-review-batches.png", "Review batches", "Summarises the concept candidates into human-scale review batches before the full audit tables appear.", "The batches are agent-proposed triage, not final concept decisions.", artifacts)

    fig = _community_count_stability_figure(stage8_result)
    _save_figure(fig, "04-community-count-stability.png", "Community count stability", "Makes the number of raw Louvain communities explicit and shows whether the count changes under nearby resolution settings.", "This describes raw diagnostic communities, not validated topics.", artifacts)

    fig = _community_composition_figure(stage7_result.get("community_review_summary", pd.DataFrame()))
    _save_figure(fig, "05-community-composition.png", "Community composition", "Shows the contents of each raw community directly, instead of relying on the circular network layout.", "Community contents are agent-triaged candidates and require human review.", artifacts)

    # The full page-community matrix is still written to CSV by _page_community_matrix(), but it is no longer exported as a review figure.
    # In larger samples it becomes a wall of page rows and stops functioning as a supervision surface.
    _page_community_matrix(stage7_result)

    # Topic count comparison is intentionally not exported as a graph.
    # Counts alone do not tell the human reviewer what the communities contain.

    fig = _domain_cap_sensitivity_figure(stage8_result)
    _save_figure(fig, "08-domain-cap-sensitivity.png", "Domain cap sensitivity", "Compares the uncapped SERP with post-hoc cap-3 and cap-1 views to test whether evidence depends on repeated same-domain results.", "Capped views are sensitivity checks; the uncapped view remains the primary SERP record.", artifacts)

    _save_permutation_networks(stage8_result, artifacts)

    raw_graph = stage7_result["raw_graph"]
    raw_nodes = stage7_result["raw_nodes"].copy()
    raw_communities = _raw_graph_communities(raw_graph)
    raw_community_labels = _write_raw_community_labels(stage7_result, raw_communities)
    community_lookup = {node: community_id for community_id, community in enumerate(raw_communities, start=1) for node in community}
    if raw_graph.number_of_nodes():
        candidate_scores = raw_nodes.copy()
        if "concept" not in candidate_scores.columns:
            candidate_scores["concept"] = list(raw_graph.nodes())[: len(candidate_scores)]
        for column in ["page_support", "domain_support", "review_priority"]:
            if column not in candidate_scores.columns:
                candidate_scores[column] = 0
        candidate_scores = candidate_scores[candidate_scores["concept"].isin(raw_graph.nodes())].copy()
        candidate_scores["page_support"] = pd.to_numeric(candidate_scores["page_support"], errors="coerce").fillna(0)
        candidate_scores["domain_support"] = pd.to_numeric(candidate_scores["domain_support"], errors="coerce").fillna(0)
        candidate_scores["review_priority"] = pd.to_numeric(candidate_scores["review_priority"], errors="coerce").fillna(0)
        filtered_candidates = candidate_scores[candidate_scores["page_support"] >= RAW_GRAPH_MIN_PAGE_SUPPORT].copy()
        if filtered_candidates.empty:
            filtered_candidates = candidate_scores.copy()
        filtered_candidates = filtered_candidates.sort_values(["review_priority", "domain_support", "page_support"], ascending=False).head(RAW_GRAPH_REVIEW_NODE_LIMIT)
        review_nodes = [node for node in filtered_candidates["concept"].tolist() if node in raw_graph]
        review_graph = raw_graph.subgraph(review_nodes).copy()
        if review_graph.number_of_edges() == 0 and len(review_nodes) > 1:
            expanded = set(review_nodes)
            for node in review_nodes[: min(25, len(review_nodes))]:
                neighbours = sorted(raw_graph[node].items(), key=lambda item: item[1].get("weight", 0), reverse=True)[:3]
                expanded.update(neighbour for neighbour, _ in neighbours)
            expanded = list(expanded)[: RAW_GRAPH_REVIEW_NODE_LIMIT]
            review_graph = raw_graph.subgraph(expanded).copy()
        fig, ax = plt.subplots(figsize=(16, 11))
        if review_graph.number_of_nodes() > 0:
            k_value = 1.15 / max(1, np.sqrt(review_graph.number_of_nodes() / 20))
            pos = nx.spring_layout(review_graph, seed=CLUSTER_RANDOM_STATE, weight="weight", k=k_value, iterations=120)
            widths = [0.15 + 0.35 * float(data.get("weight", 1)) for _, _, data in review_graph.edges(data=True)]
            nx.draw_networkx_edges(review_graph, pos, ax=ax, width=widths, alpha=0.12, edge_color="#64748b")
            ordered = candidate_scores.set_index("concept").reindex(list(review_graph.nodes())).fillna(0)
            sizes = 170 + 85 * ordered.get("page_support", pd.Series(1, index=ordered.index)).astype(float).to_numpy()
            colours = [community_lookup.get(node, 0) for node in review_graph.nodes()]
            scatter = ax.scatter([pos[node][0] for node in review_graph.nodes()], [pos[node][1] for node in review_graph.nodes()], s=sizes, c=colours, cmap="tab20", alpha=0.82, edgecolors="#111827", linewidths=0.45)
            labels = {node: str(node) for node in review_graph.nodes()}
            nx.draw_networkx_labels(review_graph, pos, labels=labels, ax=ax, font_size=7, font_color="#111827", bbox={"boxstyle": "round,pad=0.14", "fc": "white", "ec": "#e5e7eb", "alpha": 0.82})
            colorbar = fig.colorbar(scatter, ax=ax, shrink=0.72)
            colorbar.set_label("Raw diagnostic community")
        ax.set_title(f"Raw phrase co-occurrence diagnostic: top {review_graph.number_of_nodes()} review nodes, min page support {RAW_GRAPH_MIN_PAGE_SUPPORT}")
        ax.text(0.01, 0.01, f"Filtered from {raw_graph.number_of_nodes()} raw nodes / {raw_graph.number_of_edges()} raw edges. This is not a validated topic map.", transform=ax.transAxes, fontsize=9, color="#475569")
        ax.axis("off")
    else:
        fig = _empty_figure("Raw phrase graph diagnostic", "No raw phrase graph was produced.")
    _save_figure(fig, "09-raw-phrase-clique-diagnostic.png", "Raw phrase clique diagnostic", f"Shows a support/priority-limited review subgraph rather than every raw candidate. Default limit: {RAW_GRAPH_REVIEW_NODE_LIMIT} nodes with minimum page support {RAW_GRAPH_MIN_PAGE_SUPPORT}.", "This is a labelled diagnostic, not a validated semantic network. The cutoff improves reviewability but hides lower-priority raw nodes; the full graph remains available in private data outputs.", artifacts)

    fig = _community_label_figure(raw_community_labels)
    _save_figure(fig, "10-raw-community-labels.png", "Raw community labels", "Lists the candidate phrases inside each visible raw diagnostic community so the network blobs can be reviewed without relying only on spatial layout.", "Communities are produced from raw co-occurrence structure and may reflect page-local extraction artefacts, boilerplate, or source overlap.", artifacts)

    if not raw_nodes.empty:
        fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
        raw_nodes["page_support"].plot(kind="hist", bins=min(20, max(3, raw_nodes["page_support"].nunique())), ax=axes[0], color="#2563eb")
        axes[0].set_title("Raw phrase page support")
        axes[0].set_xlabel("Pages")
        raw_nodes["domain_support"].plot(kind="hist", bins=min(20, max(3, raw_nodes["domain_support"].nunique())), ax=axes[1], color="#7c3aed")
        axes[1].set_title("Raw phrase domain support")
        axes[1].set_xlabel("Domains")
        fig.suptitle("Raw candidate support distributions")
    else:
        fig = _empty_figure("Raw candidate support", "No raw candidate table was available.")
    _save_figure(fig, "11-raw-support-distributions.png", "Raw support distributions", "Shows whether extracted phrase candidates recur across pages and domains or mostly appear once.", "Support counts are recurrence signals, not proof of conceptual validity.", artifacts)

    fig = _draw_substantive_graph(stage7_result)
    if fig is not None:
        _save_figure(fig, "12-substantive-concept-graph.png", "Evidence-filtered concept graph", "Shows the graph that remains after human-reviewed concepts and independent recurrence thresholds are applied.", "Even retained edges are relationship candidates unless their evidence windows are inspected.", artifacts)

    accepted = stage7_result["provenance"] if "provenance" in stage7_result else pd.DataFrame()
    evidence_label = "reviewed"
    if not accepted.empty and {"canonical_concept", "page_id"}.issubset(accepted.columns):
        top_concepts = accepted["canonical_concept"].value_counts().head(18).index.tolist()
        sub = accepted[accepted["canonical_concept"].isin(top_concepts)].copy()
        concept_column = "canonical_concept"
    else:
        # Human supervision needs downstream context before final approval.
        # Build a provisional page-concept evidence view from the highest-priority pending candidates.
        evidence_label = "provisional pending"
        ledger_for_graph = stage7_result.get("review_ledger", pd.DataFrame()).copy()
        raw_for_graph = stage7_result.get("raw_candidates", pd.DataFrame()).copy()
        if not ledger_for_graph.empty and not raw_for_graph.empty and {"proposed_canonical", "review_priority"}.issubset(ledger_for_graph.columns):
            top_terms = ledger_for_graph.sort_values("review_priority", ascending=False)["proposed_canonical"].dropna().astype(str).head(12).tolist()
            top_terms_lower = [term.lower() for term in top_terms]
            sub = raw_for_graph[raw_for_graph["raw_surface"].astype(str).str.lower().isin(top_terms_lower)].copy() if "raw_surface" in raw_for_graph.columns else pd.DataFrame()
            if not sub.empty:
                sub["canonical_concept"] = sub["raw_surface"].astype(str)
                concept_column = "canonical_concept"
            else:
                concept_column = "canonical_concept"
        else:
            sub = pd.DataFrame()
            concept_column = "canonical_concept"

    if not sub.empty and {concept_column, "page_id"}.issubset(sub.columns):
        graph = nx.Graph()
        domain_lookup = dict(zip(sub.get("page_id", pd.Series(dtype=str)), sub.get("domain", pd.Series(dtype=str)))) if "domain" in sub.columns else {}
        for row in sub.itertuples(index=False):
            page_id = str(getattr(row, "page_id"))
            concept = str(getattr(row, concept_column))
            page_label = str(domain_lookup.get(page_id, page_id))[:32]
            graph.add_node("page:" + page_id, node_type="page", label=page_label)
            graph.add_node("concept:" + concept, node_type="concept", label=concept)
            graph.add_edge("page:" + page_id, "concept:" + concept)
        fig, ax = plt.subplots(figsize=(13, 8.5))
        pos = nx.spring_layout(graph, seed=CLUSTER_RANDOM_STATE, k=0.85)
        page_nodes = [n for n, d in graph.nodes(data=True) if d.get("node_type") == "page"]
        concept_nodes = [n for n, d in graph.nodes(data=True) if d.get("node_type") == "concept"]
        nx.draw_networkx_edges(graph, pos, ax=ax, alpha=0.18, edge_color="#64748b")
        nx.draw_networkx_nodes(graph, pos, nodelist=page_nodes, node_color="#94a3b8", node_size=140, ax=ax, label="Pages")
        nx.draw_networkx_nodes(graph, pos, nodelist=concept_nodes, node_color="#16a34a", node_size=330, ax=ax, label="Concept candidates")
        concept_labels = {node: graph.nodes[node].get("label", node).replace("concept:", "") for node in concept_nodes}
        page_labels = {node: graph.nodes[node].get("label", node).replace("page:", "") for node in page_nodes if graph.degree(node) > 1}
        nx.draw_networkx_labels(graph, pos, labels=concept_labels, font_size=7, ax=ax, bbox={"boxstyle": "round,pad=0.12", "fc": "white", "ec": "none", "alpha": 0.72})
        nx.draw_networkx_labels(graph, pos, labels=page_labels, font_size=6, ax=ax, font_color="#475569")
        ax.set_title(f"Page-concept evidence graph ({evidence_label} concepts): concept labels are shown")
        ax.legend(loc="best")
        ax.axis("off")
    else:
        fig = _empty_figure("Page-concept evidence graph", "No page-concept evidence graph can be drawn from reviewed or high-priority provisional candidates.")
    _save_figure(fig, "13-page-concept-evidence-graph.png", "Page-concept evidence graph", "Links concept candidates back to the pages that support them, so review decisions have downstream context before final semantic claims are made.", "This graph is provisional when concepts are pending. It supports supervision; it is not yet a validated semantic claim.", artifacts)

    representation_items = list(stage7_result.get("representations", {}).items())
    if representation_items:
        cols = min(3, len(representation_items))
        rows = int(np.ceil(len(representation_items) / cols))
        fig, axes = plt.subplots(rows, cols, figsize=(5.2 * cols, 4.4 * rows), squeeze=False)
        for ax in axes.ravel():
            ax.axis("off")
        for ax, (name, matrix) in zip(axes.ravel(), representation_items):
            coords = PCA(n_components=2, random_state=CLUSTER_RANDOM_STATE).fit_transform(matrix)
            labels = stage7_result["clusters"].get(f"{name}_cluster", pd.Series([0] * len(coords))).to_numpy()
            ax.scatter(coords[:, 0], coords[:, 1], c=labels, cmap="tab10", s=60, alpha=0.82)
            ax.set_title(f"{name}: page representation projection")
            ax.set_xlabel("PC1")
            ax.set_ylabel("PC2")
            ax.axis("on")
        fig.tight_layout()
    else:
        fig = _empty_figure("Representation projection", "No embedding or TF-IDF representations were available.")
    _save_figure(fig, "14-representation-projections.png", "Representation projections", "Compares how TF-IDF and transformer representations arrange the same fetched pages in two dimensions.", "PCA is a lossy projection; visible proximity is a prompt for inspection, not a proof of semantic identity.", artifacts)

    representation_pairs = stage8_result.get("representation_pairs", pd.DataFrame()).copy()
    if not representation_pairs.empty:
        if "pair" not in representation_pairs.columns and {"left", "right"}.issubset(representation_pairs.columns):
            representation_pairs["pair"] = representation_pairs["left"].astype(str) + " vs " + representation_pairs["right"].astype(str)
        fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
        representation_pairs.plot(kind="bar", x="pair", y="top5_neighbour_jaccard", ax=axes[0], color="#0f766e", legend=False)
        axes[0].set_title("Top-5 neighbour overlap")
        axes[0].set_ylim(0, 1)
        representation_pairs.plot(kind="bar", x="pair", y="cluster_ari", ax=axes[1], color="#be123c", legend=False)
        axes[1].set_title("Cluster agreement, adjusted Rand index")
        axes[1].set_ylim(-0.1, 1)
        for ax in axes:
            ax.tick_params(axis="x", rotation=35)
        fig.tight_layout()
    else:
        fig = _empty_figure("Representation agreement", "Model agreement diagnostics were not available.")
    _save_figure(fig, "15-representation-agreement.png", "Representation agreement", "Shows whether different Python representation methods produce similar neighbours and clusters.", "Agreement across models is a stability signal, not a guarantee that the resulting categories are useful for SEO decisions.", artifacts)

    extraction = stage8_result.get("extraction", pd.DataFrame())
    if not extraction.empty and "token_jaccard" in extraction.columns:
        fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))
        extraction["token_jaccard"].plot(kind="hist", bins=min(12, max(3, extraction["token_jaccard"].nunique())), ax=axes[0], color="#0f766e")
        axes[0].set_title("Token overlap after extraction cleaning")
        axes[0].set_xlabel("Token Jaccard")
        axes[0].set_xlim(0, 1.05)
        ratio_column = "length_ratio_trafilatura_to_beautifulsoup"
        if ratio_column in extraction.columns:
            extraction[ratio_column].plot(kind="hist", bins=min(12, max(3, extraction[ratio_column].nunique())), ax=axes[1], color="#7c3aed")
            axes[1].axvline(1.0, color="#111827", linewidth=1, alpha=0.45)
            axes[1].set_title("Extracted length ratio")
            axes[1].set_xlabel("Trafilatura / BeautifulSoup characters")
        else:
            axes[1].axis("off")
            axes[1].text(0.5, 0.5, "No length-ratio column available", ha="center", va="center")
        fig.suptitle("Extraction sensitivity: visible summary of text-preparation effects")
        fig.tight_layout()
    elif not extraction.empty and {"token_jaccard_to_baseline", "char_jaccard_to_baseline"}.issubset(extraction.columns):
        fig, ax = plt.subplots(figsize=(10, 5))
        extraction[["token_jaccard_to_baseline", "char_jaccard_to_baseline"]].plot(kind="box", ax=ax)
        ax.set_title("Extraction sensitivity against baseline text")
        ax.set_ylabel("Jaccard similarity")
        ax.set_ylim(0, 1.05)
    else:
        fig = _empty_figure("Extraction sensitivity", "Extraction sensitivity diagnostics were not available or did not contain recognised overlap columns.")
    _save_figure(fig, "16-extraction-sensitivity.png", "Extraction sensitivity", "Tests how much the page text changes when boilerplate and low-information lines are removed.", "This measures text-preparation sensitivity only; it does not establish whether the cleaned text is the best possible representation.", artifacts)

    write_json(ANALYSIS_DIR / "visualisation_manifest.json", {"snapshot_id": SNAPSHOT_ID, "figures": artifacts})
    write_checkpoint("stage8_visualisations", "complete", figure_count=len(artifacts))
    return artifacts


visualisation_artifacts = []
if stage8:
    visualisation_artifacts = generate_stage8_visualisations(stage7, stage8, acquisition)
    display(pd.DataFrame(visualisation_artifacts))
else:
    print("Stage 8 visualisations require completed Stage 7 and Stage 8 results.")
````

````python
import base64
import html
import nbformat

EXPORT_REVIEW_ARTIFACTS = os.getenv("SEO_LAB_EXPORT_REVIEW_ARTIFACTS", "true").strip().lower() == "true"


SUPERVISION_COLUMNS = [
    "proposed_canonical", "suggested_action", "suggested_reason",
    "page_support", "domain_support", "raw_community",
    "target_term_match", "raw_surfaces", "decision",
]
COMPACT_BATCH_COLUMNS = ["review_batch", "candidate_count", "top_candidates", "next_human_action"]
COMPACT_COMMUNITY_COLUMNS = ["raw_community", "candidate_count", "review_first", "target_terms", "likely_rejects", "human_question"]
TOPIC_COMPARISON_COLUMNS = ["method", "unit_clustered", "groups_detected", "status", "limitation"]
DOMAIN_CAP_COLUMNS = ["view", "domain_cap", "query_hits", "unique_domains", "unique_pages", "unique_raw_candidates", "top_candidates"]
METHOD_LABEL_COLUMNS = ["permutation", "group_id", "label_terms", "human_provisional_label", "page_count", "domain_count", "status", "limitation"]
ASSESSMENT_COLUMNS = ["permutation", "codex_assessment", "review_prompt"]


def _display_frame(frame, maximum_rows=40, columns=None):
    if frame is None:
        return pd.DataFrame()
    display_frame = frame.copy() if isinstance(frame, pd.DataFrame) else pd.DataFrame(frame)
    if columns:
        available = [column for column in columns if column in display_frame.columns]
        display_frame = display_frame[available]
    return display_frame.head(maximum_rows).copy()


def review_output_cell(title, frame, maximum_rows=40, columns=None):
    display_frame = _display_frame(frame, maximum_rows, columns=columns)
    note = f"Showing first {len(display_frame)} rows. Full private outputs remain in {OUTPUT_DIR}."
    cell = nbformat.v4.new_code_cell(f"# Rendered review output: {title}")
    cell["execution_count"] = None
    cell["outputs"] = [nbformat.v4.new_output(
        output_type="display_data",
        data={
            "text/html": f"<p><em>{html.escape(note)}</em></p>{display_frame.to_html(index=False, escape=True)}",
            "text/plain": note + "\n" + display_frame.to_string(index=False),
        },
        metadata={},
    )]
    return cell


def review_image_cell(title, source_path):
    source_path = Path(source_path)
    cell = nbformat.v4.new_code_cell(f"# Rendered review figure: {title}")
    cell["execution_count"] = None
    if source_path.exists():
        image_bytes = base64.b64encode(source_path.read_bytes()).decode("ascii")
        cell["outputs"] = [nbformat.v4.new_output(
            output_type="display_data",
            data={"image/png": image_bytes, "text/plain": f"Figure: {title}"},
            metadata={},
        )]
    else:
        cell["outputs"] = [nbformat.v4.new_output(output_type="stream", name="stdout", text=f"Missing figure: {source_path}\n")]
    return cell


def _markdown_to_html(markdown_text):
    lines = markdown_text.splitlines()
    html_lines = []
    paragraph = []
    list_items = []

    def flush_paragraph():
        nonlocal paragraph
        if paragraph:
            html_lines.append("<p>" + html.escape(" ".join(paragraph)) + "</p>")
            paragraph = []

    def flush_list():
        nonlocal list_items
        if list_items:
            html_lines.append("<ul>" + "".join(f"<li>{html.escape(item)}</li>" for item in list_items) + "</ul>")
            list_items = []

    for raw_line in lines:
        line = raw_line.rstrip()
        stripped = line.strip()
        if not stripped:
            flush_paragraph()
            flush_list()
            continue
        if stripped.startswith("### "):
            flush_paragraph(); flush_list()
            html_lines.append("<h3>" + html.escape(stripped[4:]) + "</h3>")
        elif stripped.startswith("## "):
            flush_paragraph(); flush_list()
            html_lines.append("<h2>" + html.escape(stripped[3:]) + "</h2>")
        elif stripped.startswith("# "):
            flush_paragraph(); flush_list()
            html_lines.append("<h1>" + html.escape(stripped[2:]) + "</h1>")
        elif stripped.startswith("> "):
            flush_paragraph(); flush_list()
            html_lines.append("<blockquote>" + html.escape(stripped[2:]) + "</blockquote>")
        elif stripped.startswith("- "):
            flush_paragraph()
            list_items.append(stripped[2:])
        else:
            flush_list()
            paragraph.append(stripped)
    flush_paragraph()
    flush_list()
    return "\n".join(html_lines)


def _details_table(title, frame, maximum_rows=80, open_section=False, columns=None, note=None):
    display_frame = _display_frame(frame, maximum_rows, columns=columns)
    table = display_frame.to_html(index=False, escape=True, classes="data-table") if not display_frame.empty else "<p><em>No rows.</em></p>"
    open_attr = " open" if open_section else ""
    note_html = f"<p class=\'section-note\'>{html.escape(note)}</p>" if note else ""
    return f"<details{open_attr}><summary>{html.escape(title)} ({len(display_frame)} shown)</summary>{note_html}<div class=\'table-wrap\'>{table}</div></details>"


def _review_dashboard_html(review_summary, review_batch_summary):
    row = review_summary.iloc[0].to_dict() if not review_summary.empty else {}
    batch_lookup = review_batch_summary.set_index('review_batch')['candidate_count'].to_dict() if not review_batch_summary.empty and 'review_batch' in review_batch_summary.columns else {}
    high_priority = row.get('high_priority_candidates')
    if high_priority in (None, '', 'None'):
        high_priority = batch_lookup.get('batch_1_high_priority_target_aligned', 0)
    cards = [
        ('Pending review', row.get('pending_concept_review', 0), 'Audit count, not the human queue size'),
        ('First decisions', high_priority, 'Small queue to inspect first'),
        ('Substantive graph', 'ready' if row.get('substantive_graph_ready') else 'provisional context first', 'Do not block all context before review'),
        ('Raw edge risk', row.get('raw_single_page_edge_ratio', ''), 'Share of raw edges with one-page support'),
    ]
    card_html = ''.join(
        f"<div class='dashboard-card'><div class='dashboard-label'>{html.escape(str(label))}</div><div class='dashboard-value'>{html.escape(str(value))}</div><div class='dashboard-note'>{html.escape(str(note))}</div></div>"
        for label, value, note in cards
    )
    return "<section class='dashboard'><h2>Review dashboard</h2><div class='dashboard-grid'>" + card_html + "</div></section>"


def _supervision_principle_html():
    return """
<section class='supervision-principle'>
<h2>Audit trail versus supervision surface</h2>
<p><strong>Correction:</strong> the full audit trail is not the human review interface. The audit trail preserves everything Codex did; the supervision surface shows only the decisions the human can meaningfully make right now.</p>
<ul>
<li><strong>Codex needs:</strong> complete inputs, code, assumptions, intermediate outputs, failed renderings, diagnostics and revisions.</li>
<li><strong>The human supervisor needs:</strong> compact decisions, labels, recommendations, consequences, and provisional downstream context.</li>
</ul>
<blockquote>Human supervision should not be requested on raw extraction exhaust before the agent has made the consequences of those extraction choices visible.</blockquote>
</section>
"""


def _figure_role(title):
    if title.startswith("Permutation: "):
        return "supervision"
    supervision_titles = {
        "Acquisition coverage",
        "Review batches",
        "Raw community labels",
        "Domain concentration",
        "Community count stability",
        "Community composition",
        "Topic structure comparison",
        "Domain cap sensitivity",
        "Page-concept evidence graph",
        "Representation projections",
        "Representation agreement",
        "Extraction sensitivity",
    }
    return "supervision" if title in supervision_titles else "audit"


def _figure_html(figure, collapsible=False, open_section=False):
    figure_path = assert_private_output_path(OUTPUT_DIR / figure["path"])
    if not figure_path.exists():
        return f"<section><h2>{html.escape(figure['title'])}</h2><p><em>Missing figure: {html.escape(str(figure_path))}</em></p></section>"
    encoded = base64.b64encode(figure_path.read_bytes()).decode("ascii")
    caption = html.escape(figure.get("caption", ""))
    limitation = html.escape(figure.get("limitation", ""))
    return (
        f"<section><h2>{html.escape(figure['title'])}</h2>"
        f"<img src='data:image/png;base64,{encoded}' alt='{html.escape(figure['title'])}'>"
        f"<p>{caption}</p><p><strong>Limitation:</strong> {limitation}</p></section>"
    )




def _preliminary_permutation_findings_markdown():
    return "## Preliminary finding: what the permutations demonstrate\n\nThe first useful public interpretation of the method-label permutation layer is not that one model is the winner. The more defensible finding is that each permutation exposes a different kind of structure in the same search-result environment.\n\n| Permutation | Biggest demonstrated strength | Biggest demonstrated weakness | Practical SEO implication |\n|---|---|---|---|\n| NetworkX-Louvain x TF-IDF | Best discovery map. It exposed many distinct SERP regions: Python SEO automation, Selenium, RankBrain, internal linking, content gaps, keyword clustering, BERT, RoBERTa, spaCy, NLP, and NetworkX. | Noisy and visually dense. It treats page-wide co-occurrence as structure, so artefacts and useful concepts can appear together. | Use it to find the search territory, not to make final claims. It shows what topics the site may need to address. |\n| NetworkX-Louvain x c-TF-IDF | Best community-labelling improvement over raw TF-IDF. It made the Louvain communities easier to interpret by emphasising terms distinctive to each community. | It inherits Louvain's structure. If the community split is questionable, c-TF-IDF only labels that questionable split more clearly. | Good for turning graph communities into tentative article or topic labels, but still needs human review. |\n| KMeans-BERT x TF-IDF | Best at grouping pages by broader semantic similarity. It showed that some topics belong together at a higher conceptual level. | Too smooth. It merged areas the human reviewer may want separated, such as Semantic SEO, NLP, BERT, RankBrain and entities. | Useful for seeing broad search-intent regions, but risky for detailed editorial planning because it can flatten important distinctions. |\n| KMeans-RoBERTa x TF-IDF | Best practical middle ground among embedding methods. It separated several useful SEO/content areas more cleanly than BERT: internal linking, content gaps, NetworkX, spaCy, keyword clustering, automation, and commercial SEO. | Still depends on fixed `k` clustering, meaning the number of topics is imposed before interpretation. | Strong candidate for page and article planning, especially when the goal is readable topic separation rather than pure network topology. |\n| BERTopic-HDBSCAN x c-TF-IDF | Best at finding natural dense topic families without forcing every page into a neat topic. It reduced the field to major clusters: internal links, spaCy, content gaps, Python SEO automation, and Semantic SEO/NLP/clustering. | It compressed the search space heavily. Smaller but strategically useful topics like BERT, RoBERTa or NetworkX can disappear into larger clusters. | Useful for identifying dominant public-facing topic pillars, but not enough on its own for a full Semantic SEO/Python strategy. |\n| Human labels x agent critique | Best interpretive layer. It preserved distinctions the models blurred and turned computed clusters into meaningful SEO narratives. | Subjective and dependent on review quality. It can overfit to the project's thesis if not challenged by the data. | This is the strategic layer. The models expose structure; human supervision decides what matters, what is noise, and what should become content. |\n\nThe notebook does not use these permutations to find a single correct topic model. It uses them to expose different representations of the same SERP space. Network methods reveal relationship structure, embedding methods reveal semantic proximity, BERTopic reveals dense topic families, and human-agent review turns those outputs into supervised SEO strategy.\n\nThe strongest preliminary finding is that the SERP space is not one topic. It is a fragmented network of implementation methods, SEO tasks, model explanations and commercial content. The page's opportunity is to make those fragments coherent through a visible, supervised Semantic SEO Python workflow.\n\n## Preliminary finding: topic narrative by permutation\n\nThe topic results should be read as overlapping representations of the same search environment rather than as six independent truths. Each permutation changes what becomes visible: graph methods emphasise co-occurring search vocabulary, embedding methods emphasise page-level semantic proximity, BERTopic/HDBSCAN compresses the space into dense topic families, and the human-label layer converts provisional clusters into supervised editorial meaning.\n\n**NetworkX-Louvain x TF-IDF** produced the most granular territory map. It separates the SERP space into many recognisable regions: Python SEO automation, Selenium-based page SEO, machine-learning SEO automation, Semantic SEO around RankBrain, entities and Claude, internal linking, competitor and content-gap analysis, commercial SEO services, keyword clustering, RoBERTa and embeddings, spaCy, BERT, NLP, NetworkX, ecommerce landing-page analysis, and Python/NLP/Semantic SEO. Its value is breadth. It makes the fragmentation of the search space visible, but it also carries the most noise because raw TF-IDF phrases and page-wide co-occurrence can turn accidental proximity into graph structure.\n\n**NetworkX-Louvain x c-TF-IDF** keeps the same underlying Louvain community structure but improves the labels. Its main contribution is interpretive sharpening: terms distinctive to each community become easier to see, which makes the graph more useful as a topic-review surface. It does not independently validate the communities. If Louvain has split or merged the space badly, c-TF-IDF only gives a cleaner name to that questionable split.\n\n**KMeans-BERT x TF-IDF** reads the pages at a broader semantic level. It tends to merge regions that are conceptually related, such as Semantic SEO, NLP, BERT, RankBrain and entities. This is useful for seeing large search-intent zones, but it is less useful for detailed editorial planning because the smoothing can hide distinctions that matter to the site strategy. BERT therefore behaves more like a broad map of semantic neighbourhoods than a fine-grained topic architecture.\n\n**KMeans-RoBERTa x TF-IDF** produced a more practical middle ground among the embedding approaches. It preserved several editorially useful regions more clearly: internal linking, content gaps, NetworkX, spaCy, keyword clustering, automation, commercial SEO, Semantic SEO/NLP/BERT, and RoBERTa/embedding material. Its weakness is that KMeans still imposes a fixed number of clusters before interpretation begins. The clusters are useful, but the notebook should not pretend their count is naturally discovered.\n\n**BERTopic-HDBSCAN x c-TF-IDF** compressed the space into five dense topic families: internal linking and site structure; spaCy implementation and pipeline material; content-gap and competitor analysis; Python SEO automation; and a broad Semantic SEO/NLP/keyword-clustering/RankBrain/BERT region. Its value is that it does not force every page into a neat topic. Its weakness is compression: smaller strategic regions such as BERT, RoBERTa, NetworkX and some automation subtopics can be absorbed into larger families.\n\n**Human labels x agent critique** is the richest interpretive layer because it preserves distinctions the computational methods blur. The human labels identified concrete content-strategy meanings: automating page SEO with Selenium, machine-learning SEO automation, Claude and RankBrain in Semantic SEO, internal linking strategy, competitor gap analysis, commercial SEO offerings, keyword clustering, RoBERTa, spaCy, BERT, NLP, NetworkX, ecommerce landing-page analysis, and BERTopic-based Python/NLP/Semantic SEO. The risk is subjectivity. These labels must remain challengeable, because a human-agent interpretation can overfit the project's desired thesis if it is not checked against the computed evidence.\n\nAcross the permutations, the recurring topic system is not a single `semantic seo python` topic. It is a network of implementation methods, SEO tasks, model-explanation pages and commercial offerings. The recurring regions are Python SEO automation; Semantic SEO, NLP, RankBrain and entities; keyword clustering; content-gap and competitor analysis; internal linking and site structure; spaCy; BERT; RoBERTa and embeddings; NetworkX and graph analysis; and commercial SEO services.\n\nThe practical strategic finding is that the existing SERP space is fragmented. The notebook can occupy a stronger position by showing how these fragments become one supervised Semantic SEO Python workflow: Python collects and measures, NLP and embeddings produce candidate representations, NetworkX and clustering expose structure, BERTopic and c-TF-IDF test topic families, Codex interprets and critiques the outputs, and human supervision decides which interpretations are defensible enough to publish.\n"


def _permutation_assessment_markdown(stage8_result):
    findings_markdown = _preliminary_permutation_findings_markdown()
    assessments = stage8_result.get("codex_permutation_assessments", pd.DataFrame())
    if assessments is None or assessments.empty:
        return findings_markdown
    parts = [findings_markdown, "", "## Codex permutation assessments", "", "These are concise agent interpretations placed after the visuals for human challenge, correction, or approval."]
    for _, row in assessments.iterrows():
        parts.extend([
            "",
            f"### {row.get('permutation', '')}",
            "",
            str(row.get('codex_assessment', '') or '').strip(),
            "",
            f"**Human review prompt:** {str(row.get('review_prompt', '') or '').strip()}",
        ])
    return "\n".join(parts)

def _sort_if_columns(frame, columns, ascending=False):
    if frame is None or not isinstance(frame, pd.DataFrame) or frame.empty:
        return pd.DataFrame() if frame is None else frame
    available = [column for column in columns if column in frame.columns]
    if not available:
        return frame
    if isinstance(ascending, list):
        ascending = ascending[:len(available)]
    return frame.sort_values(available, ascending=ascending)

def export_private_review_artifacts(stage7_result, stage8_result):
    if stage7_result is None or stage8_result is None:
        raise RuntimeError("Private review export requires completed Stages 7 and 8.")

    diagnostics = stage8_result["diagnostics"]
    review_dir = assert_private_output_path(OUTPUT_DIR / "review")
    review_dir.mkdir(parents=True, exist_ok=True)
    executed_notebook_path = review_dir / f"semantic-seo-lab-{SNAPSHOT_ID}-review.ipynb"
    html_path = review_dir / f"semantic-seo-lab-{SNAPSHOT_ID}-review.html"
    export_manifest_path = review_dir / "review_artifacts.json"

    review_notebook = nbformat.read(NOTEBOOK_PATH, as_version=4)
    review_notebook["metadata"]["semantic_seo_lab_review"] = {
        "snapshot_id": SNAPSHOT_ID,
        "private_review_only": True,
        "source_notebook": str(NOTEBOOK_PATH),
    }

    summary = stage7_result["summary"]
    review_gate = diagnostics["concept_review_gate"]
    review_summary = pd.DataFrame([{
        "snapshot_id": SNAPSHOT_ID,
        "documents": summary.get("documents"),
        "raw_phrase_nodes": summary.get("raw_phrase_nodes"),
        "raw_phrase_edges": summary.get("raw_phrase_edges"),
        "substantive_concept_nodes": summary.get("substantive_concept_nodes"),
        "substantive_concept_edges": summary.get("substantive_concept_edges"),
        "pending_concept_review": review_gate.get("pending_candidates"),
        "accepted_or_renamed": review_gate.get("accepted_or_renamed"),
        "high_priority_candidates": review_gate.get("high_priority_candidates"),
        "substantive_graph_ready": review_gate.get("substantive_graph_ready"),
        "raw_single_page_edge_ratio": diagnostics["raw_phrase_graph_reliability"].get("single_page_edge_ratio"),
        "substantive_single_page_edge_ratio": diagnostics["substantive_graph_reliability"].get("single_page_edge_ratio"),
    }])

    intro = (
        "# Private run review\n\n"
        f"This review records snapshot `{SNAPSHOT_ID}`. It is a private derivative for inspection before any public article or GitHub upload.\n\n"
        "The review is now split into a human supervision surface and a full audit appendix. The supervision surface is compact and decision-oriented; the audit appendix preserves the raw tables and diagnostic figures.\n\n"
        f"Current review instruction: {review_gate.get('next_required_action')}"
    )
    review_notebook["cells"].append(nbformat.v4.new_markdown_cell(intro))

    priority_review = _sort_if_columns(stage7_result["review_ledger"], ["review_priority"], ascending=False)
    batch_one = priority_review[priority_review["review_batch"].eq("batch_1_high_priority_target_aligned")].copy() if "review_batch" in priority_review.columns else pd.DataFrame()
    decision_queue = _display_frame(batch_one, maximum_rows=12, columns=SUPERVISION_COLUMNS)
    review_tables = [
        ("Run summary", review_summary, True, 20, None, "Run-level context only."),
        ("Decision queue: inspect first", decision_queue, True, 12, None, "The first human review batch. This replaces the wide full ledger as the main supervision table."),
        ("Review batch summary", stage7_result.get("review_batch_summary", pd.DataFrame()), True, 20, COMPACT_BATCH_COLUMNS, "Shows the review workload by batch; full candidate details are in the audit appendix."),
        ("Community review summary", stage7_result.get("community_review_summary", pd.DataFrame()), False, 20, COMPACT_COMMUNITY_COLUMNS, "Use this to understand what the visible raw communities might contain before deciding individual terms."),
        ("Topic structure count note", stage8_result.get("topic_comparison", pd.DataFrame()), False, 20, TOPIC_COMPARISON_COLUMNS, "Count-only comparison retained for audit; use the permutation visuals for interpretation."),
        ("Domain cap sensitivity", stage8_result.get("domain_cap_sensitivity", pd.DataFrame()), False, 20, DOMAIN_CAP_COLUMNS, "Shows what changes when repeated same-domain results are controlled after acquisition."),
    ]
    audit_tables = [
        ("Full priority review ledger", priority_review, False, 120, None, "Complete audit ledger. Not intended as the primary supervision interface."),
        ("Full concept review ledger", stage7_result["review_ledger"], False, 120, None, None),
        ("Raw phrase candidates", _sort_if_columns(stage7_result["raw_candidates"], ["page_support", "domain_support", "tfidf_score"], ascending=False), False, 120, None, None),
        ("Reviewed concept provenance", stage7_result["provenance"], False, 120, None, None),
        ("Relationship candidates", _sort_if_columns(stage7_result["relationship_candidates"], ["passes_substantive_threshold", "domain_support", "page_support"], ascending=False), False, 120, None, None),
        ("Raw graph nodes", _sort_if_columns(stage7_result["raw_nodes"], ["page_support", "domain_support"], ascending=False), False, 120, None, None),
        ("Raw graph edges", _sort_if_columns(stage7_result["raw_edges"], ["page_support", "domain_support"], ascending=False), False, 120, None, None),
        ("Substantive graph nodes", stage7_result["substantive_nodes"], False, 120, None, None),
        ("Substantive graph edges", stage7_result["substantive_edges"], False, 120, None, None),
        ("Representation agreement", stage8_result["representation_pairs"], False, 80, None, None),
        ("Method-label permutation audit", stage8_result.get("method_label_permutations", pd.DataFrame()), False, 160, METHOD_LABEL_COLUMNS, "Full permutation rows for audit; use visuals and assessments for supervision."),
        ("Codex permutation assessment audit", stage8_result.get("codex_permutation_assessments", pd.DataFrame()), False, 20, ASSESSMENT_COLUMNS, "Agent interpretation text preserved as a table for audit."),
        ("KMeans seed stability", stage8_result["seed_stability"], False, 80, None, None),
        ("Extraction sensitivity", stage8_result["extraction"], False, 80, None, None),
        ("Full topic structure comparison", stage8_result.get("topic_comparison", pd.DataFrame()), False, 80, None, None),
        ("Domain cap concept persistence", stage8_result.get("domain_cap_concept_persistence", pd.DataFrame()), False, 120, None, None),
    ]
    table_sections = review_tables + audit_tables

    for title, frame, _open_section, maximum_rows, columns, note in table_sections:
        review_notebook["cells"].append(nbformat.v4.new_markdown_cell(f"## {title}" + (f"\n\n{note}" if note else "")))
        review_notebook["cells"].append(review_output_cell(title, frame, maximum_rows=maximum_rows, columns=columns))

    for figure in visualisation_artifacts:
        figure_path = assert_private_output_path(OUTPUT_DIR / figure["path"])
        figure_markdown = f"## {figure['title']}\n\n{figure.get('caption', '')}\n\n**Limitation:** {figure.get('limitation', '')}"
        review_notebook["cells"].append(nbformat.v4.new_markdown_cell(figure_markdown))
        review_notebook["cells"].append(review_image_cell(figure["title"], figure_path))

    nbformat.write(review_notebook, executed_notebook_path)

    body = [
        "<!doctype html><html><head><meta charset='utf-8'><title>Semantic SEO Lab review</title>",
        "<style>body{font-family:system-ui,-apple-system,Segoe UI,sans-serif;line-height:1.48;margin:2rem;max-width:1180px;color:#111827}h1,h2,h3{line-height:1.15}code{background:#f3f4f6;padding:.08rem .25rem;border-radius:4px}.table-wrap{max-width:100%;overflow:auto}table{border-collapse:collapse;width:100%;font-size:.92rem;margin:1rem 0}th,td{border-bottom:1px solid #d1d5db;padding:.4rem .5rem;text-align:left;vertical-align:top}th{background:#f3f4f6}details{border:1px solid #d1d5db;border-radius:8px;padding:.75rem;margin:1rem 0}summary{font-weight:700;cursor:pointer}img{display:block;max-width:100%;height:auto;margin:1rem 0;border:1px solid #e5e7eb}blockquote{border-left:4px solid #9ca3af;margin:1rem 0;padding:.25rem 1rem;color:#374151;background:#f9fafb}.warning{background:#fff7ed;border:1px solid #fed7aa;border-radius:8px;padding:1rem}.dashboard-grid{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:.75rem;margin:1rem 0}.dashboard-card{border:1px solid #d1d5db;border-radius:8px;padding:.75rem;background:#f9fafb}.dashboard-label{font-size:.85rem;color:#4b5563}.dashboard-value{font-size:1.55rem;font-weight:700;margin:.2rem 0}.dashboard-note,.section-note{font-size:.9rem;color:#4b5563}.supervision-principle{border:1px solid #bfdbfe;background:#eff6ff;border-radius:10px;padding:1rem;margin:1rem 0}</style></head><body>",
        f"<h1>Semantic SEO Lab private review: {html.escape(SNAPSHOT_ID)}</h1>",
        "<p class='warning'><strong>Private review only.</strong> Do not publish raw snapshots, cached page text, API-bearing files, or unrelated analytics material.</p>",
        _markdown_to_html(intro),
        _review_dashboard_html(review_summary, stage7_result.get("review_batch_summary", pd.DataFrame())),
        _supervision_principle_html(),
    ]

    body.append("<h2>Human supervision surface</h2>")
    for title, frame, open_section, maximum_rows, columns, note in review_tables:
        body.append(_details_table(title, frame, maximum_rows=maximum_rows, open_section=open_section, columns=columns, note=note))

    body.append("<h2>Provisional context figures</h2>")
    body.append("<p class='section-note'>These figures are shown before final approval because they help explain why a review decision matters. They are not final semantic claims.</p>")
    for figure in visualisation_artifacts:
        if _figure_role(figure.get("title", "")) == "supervision":
            body.append(_figure_html(figure))

    assessment_markdown = _permutation_assessment_markdown(stage8_result)
    if assessment_markdown:
        body.append(_markdown_to_html(assessment_markdown))

    body.append("<h2>Full audit appendix</h2>")
    body.append("<p class='section-note'>This section preserves raw exhaust, complete ledgers and diagnostic structures. It exists for auditability, not as the main supervision path.</p>")
    for title, frame, open_section, maximum_rows, columns, note in audit_tables:
        body.append(_details_table(title, frame, maximum_rows=maximum_rows, open_section=open_section, columns=columns, note=note))

    body.append("<h2>Diagnostic figures appendix</h2>")
    for figure in visualisation_artifacts:
        if _figure_role(figure.get("title", "")) != "supervision":
            body.append(_figure_html(figure, collapsible=True))

    body.append("</body></html>")
    html_path.write_text("\n".join(body), encoding="utf-8")

    manifest = {
        "snapshot_id": SNAPSHOT_ID,
        "executed_review_notebook": executed_notebook_path.name,
        "html_report": html_path.name,
        "review_directory": str(review_dir),
        "source_notebook_outputs_preserved": False,
        "publication_warning": "Review and redact before manually uploading any derivative artifact.",
        "review_design": "Human supervision surface separated from full audit appendix.",
    }
    write_json(export_manifest_path, manifest)
    write_checkpoint("stage8_review_export", "complete", **manifest)
    return manifest


review_artifacts = None
if stage8 and EXPORT_REVIEW_ARTIFACTS:
    review_artifacts = export_private_review_artifacts(stage7, stage8)
    display(pd.DataFrame([review_artifacts]))
elif stage8:
    print("Stage 8 completed. Private review-artifact export is disabled.")
else:
    print("Review artifacts are created automatically after a completed Stage 8 run.")
````

## Preliminary finding: topic narrative by permutation

The topic results should be read as overlapping representations of the same search environment rather than as six independent truths. Each permutation changes what becomes visible: graph methods emphasise co-occurring search vocabulary, embedding methods emphasise page-level semantic proximity, BERTopic/HDBSCAN compresses the space into dense topic families, and the human-label layer converts provisional clusters into supervised editorial meaning.

**NetworkX-Louvain x TF-IDF** produced the most granular territory map. It separates the SERP space into many recognisable regions: Python SEO automation, Selenium-based page SEO, machine-learning SEO automation, Semantic SEO around RankBrain, entities and Claude, internal linking, competitor and content-gap analysis, commercial SEO services, keyword clustering, RoBERTa and embeddings, spaCy, BERT, NLP, NetworkX, ecommerce landing-page analysis, and Python/NLP/Semantic SEO. Its value is breadth. It makes the fragmentation of the search space visible, but it also carries the most noise because raw TF-IDF phrases and page-wide co-occurrence can turn accidental proximity into graph structure.

**NetworkX-Louvain x c-TF-IDF** keeps the same underlying Louvain community structure but improves the labels. Its main contribution is interpretive sharpening: terms distinctive to each community become easier to see, which makes the graph more useful as a topic-review surface. It does not independently validate the communities. If Louvain has split or merged the space badly, c-TF-IDF only gives a cleaner name to that questionable split.

**KMeans-BERT x TF-IDF** reads the pages at a broader semantic level. It tends to merge regions that are conceptually related, such as Semantic SEO, NLP, BERT, RankBrain and entities. This is useful for seeing large search-intent zones, but it is less useful for detailed editorial planning because the smoothing can hide distinctions that matter to the site strategy. BERT therefore behaves more like a broad map of semantic neighbourhoods than a fine-grained topic architecture.

**KMeans-RoBERTa x TF-IDF** produced a more practical middle ground among the embedding approaches. It preserved several editorially useful regions more clearly: internal linking, content gaps, NetworkX, spaCy, keyword clustering, automation, commercial SEO, Semantic SEO/NLP/BERT, and RoBERTa/embedding material. Its weakness is that KMeans still imposes a fixed number of clusters before interpretation begins. The clusters are useful, but the notebook should not pretend their count is naturally discovered.

**BERTopic-HDBSCAN x c-TF-IDF** compressed the space into five dense topic families: internal linking and site structure; spaCy implementation and pipeline material; content-gap and competitor analysis; Python SEO automation; and a broad Semantic SEO/NLP/keyword-clustering/RankBrain/BERT region. Its value is that it does not force every page into a neat topic. Its weakness is compression: smaller strategic regions such as BERT, RoBERTa, NetworkX and some automation subtopics can be absorbed into larger families.

**Human labels x agent critique** is the richest interpretive layer because it preserves distinctions the computational methods blur. The human labels identified concrete content-strategy meanings: automating page SEO with Selenium, machine-learning SEO automation, Claude and RankBrain in Semantic SEO, internal linking strategy, competitor gap analysis, commercial SEO offerings, keyword clustering, RoBERTa, spaCy, BERT, NLP, NetworkX, ecommerce landing-page analysis, and BERTopic-based Python/NLP/Semantic SEO. The risk is subjectivity. These labels must remain challengeable, because a human-agent interpretation can overfit the project's desired thesis if it is not checked against the computed evidence.

Across the permutations, the recurring topic system is not a single `semantic seo python` topic. It is a network of implementation methods, SEO tasks, model-explanation pages and commercial offerings. The recurring regions are Python SEO automation; Semantic SEO, NLP, RankBrain and entities; keyword clustering; content-gap and competitor analysis; internal linking and site structure; spaCy; BERT; RoBERTa and embeddings; NetworkX and graph analysis; and commercial SEO services.

The practical strategic finding is that the existing SERP space is fragmented. The notebook can occupy a stronger position by showing how these fragments become one supervised Semantic SEO Python workflow: Python collects and measures, NLP and embeddings produce candidate representations, NetworkX and clustering expose structure, BERTopic and c-TF-IDF test topic families, Codex interprets and critiques the outputs, and human supervision decides which interpretations are defensible enough to publish.
